## QA-01. Annotation QA(라벨링 품질검수) 목적

### 목적

본 단계에서는 Pilot Object Labeling (파일럿 객체 라벨링) 결과의 **정확성, 완전성, 일관성**을 검증한다.

단순한 데이터 구조 오류뿐만 아니라 Bounding Box(바운딩 박스), Class Label(클래스 라벨), 
객체 속성 및 라벨링 기준의 일관성을 점검하여 향후 데이터 분석 및 모델 학습에 활용할 수 있는 품질 수준인지 평가한다. 

### 주요 검수 범위

1. **Schema Validation(스키마 검증)**
   - 필수 컴럼과 데이터 구조가 올바르게 구성되어 있는지 확인
   - Missing Value(결측값) 및 잘못된 데이터 형식 확인

2. **Bounding Box Validation(바운딩 박스 검증)**
   - `x_min < x_max`, `y_min < y_max` 조건 확인
   - 바운딩 박스 좌표가 이미지 범위를 벗어나지 않는지 확인

3. **Class Label Validation(클래스 라벨 검증)**
   - 사전에 정의된 클래스만 사용되었는지 확인
   - 오탈자 및 일관되지 않은 클래스명 탐지

4. **Object ID Validation(객체 ID 검증)**
   - 중복된 Object ID(객체 ID) 확인
   - 객체 식별 규칙이 일관되게 적용되었는지 검증

5. **Attribute Validation(속성 검증)**
   - Occlusion(가림), Truncation(잘림), Small/Distant(소형·원거리), Ambiguity(모호성) 값 검증
   - 정의되지 않는 값 또는 비정상적인 속성 조합 확인 

6. **Annotation Consistency(라벨링 일관성)**
   - 유사한 객체에 동일한 라벨링 기준이 적용되었는지 확인 
   - 추가 검토가 필요한 Edge Case(경계 사례) 식별

7. **Visual QA(시각적 품질검수)**
   - 이미지에 Bounding Box(바운딩 박스)와 Class Label(클래스 라벨)을 표시 
   - 객체 포함 범위, 박스 위치 및 클래스 정확성을 육안으로 검토 

### 기대 결과 

본 QA(품질검수)를 통해 다음 결과를 도출한다. 

- 전체 라벨링 수 및 유효/오류 라벨링 수

## QA-04. Bounding Box Validation(바운딩 박스 검증)

### 목적

각 Object Labeling(객체 라벨링)에 생성된 Bounding Box(바운딩 박스)의 
좌표와 크기가 기하학적으로 유효한지 검증한다. 

잘못된 좌표 순서, 0 이하의 Width/Height(너비/높이), 음수 좌표 및 이미지 경계를
벗어난 Bounding Box(바운딩 박스)를 탐지하여 객체 위치 정보의 품질을 평가한다. 

### 주요 검증 항목 

- `x_min < x_max` 조건 확인
- `y_min < y_max` 조건 확인
- Bounding Box Width(바운딩 박스 너비) > 0
- Bounding Box Height(바운딩 박스 높이) > 0
- Negative Coordinate(음수 좌표) 여부 확인 
- Image Boundary(이미지 경계) 초과 여부 확인
- 비정상적인 Bounding Box Size(바운딩 박스 크기) 탐지

```python
# Bounding Box Geometry(바운딩 박스 기하 구조) 검증

# Width(너비)와 Height(높이) 계산
df["bbox_width"] = df["x_max"] - df["x_min"]
df["bbox_height"] = df["y_max"] - df["y_min"]

# 잘못된 좌표 순서 및 크기 탐지
invalid_x = df["x_min"] >= df["x_max"]
invalid_y = df["y_min"] >= df["y_max"]

invalid_width = df["bbox_width"] <= 0
invalid_height = df["bbox_height"] <= 0

# 음수 좌표 탐지
negative_coordinates = (
    (df["x_min"] < 0) |
    (df["y_min"] < 0) |
    (df["x_max"] < 0) |
    (df["y_max"] < 0)
)

print("=== Bounding Box Geometry Validation ===")
print(f"Total Bounding Boxes: {len(df)}")
print(f"Invalid X Coordinates: {invalid_x.sum()}")
print(f"Invalid Y Coordinates: {invalid_y.sum()}")
print(f"Invalid Width: {invalid_width.sum()}")
print(f"Invalid Height: {invalid_height.sum()}")
print(f"Negative Coordinates: {negative_coordinates.sum()}")

geometry_issues = (
    invalid_x |
    invalid_y |
    invalid_width |
    invalid_height |
    negative_coordinates
)

print(f"\nBounding Boxes with Geometry Issues: {geometry_issues.sum()}")

if geometry_issues.sum() == 0:
    print("Result: PASS - All bounding boxes have valid geometry.")
else:
    print("Result: REVIEW - Invalid bounding box geometry detected.")
```

### QA-04-3. Annotation–Image Matching Validation (라벨링–이미지 매칭 검증)

라벨링 데이터셋에 기록된 모든 이미지가 실제 Pilot Image Directory
(파일럿 이미지 디렉터리)에 존재하는지 검증한다.

이 검증을 통해 각 Bounding Box(바운딩 박스) 라벨링이 해당 원본 이미지와
정확하게 연결되어 있는지 확인하고, 이후 Image Boundary Validation
(이미지 경계 검증)을 수행하기 위한 데이터 정합성을 확보한다.

**검증 항목**
- 라벨링 데이터에 포함된 고유 이미지 수 확인
- Pilot Image Directory(파일럿 이미지 디렉터리)의 실제 이미지 수 확인
- 라벨링 데이터와 실제 이미지 파일 간 매칭 여부 확인
- 누락된 이미지 및 라벨링에서 참조되지 않은 이미지 확인

```python
# Pilot Image Directory(파일럿 이미지 디렉터리)
pilot_dir = project_root / "data" / "pilot"

# CSV에 기록된 고유 Image Name(이미지명)
annotated_images = set(df["image_name"].unique())

# 실제 파일럿 이미지
pilot_images = {
    path.name
    for path in pilot_dir.glob("*")
    if path.suffix.lower() in {".jpg", ".jpeg", ".png"}
}

# Matching Validation(매칭 검증)
matched_images = annotated_images & pilot_images
missing_images = annotated_images - pilot_images
unreferenced_images = pilot_images - annotated_images

print("=== Annotation-Image Matching Validation ===")
print(f"Annotated Images: {len(annotated_images)}")
print(f"Pilot Image Files: {len(pilot_images)}")
print(f"Matched Images: {len(matched_images)}")
print(f"Missing Images: {len(missing_images)}")
print(f"Unreferenced Pilot Images: {len(unreferenced_images)}")

if missing_images:
    print("\nMissing Image Files:")
    for name in sorted(missing_images):
        print("-", name)

if unreferenced_images:
    print("\nUnreferenced Pilot Images:")
    for name in sorted(unreferenced_images):
        print("-", name)

if len(missing_images) == 0:
    print("\nResult: PASS - All annotated images were found.")
else:
    print("\nResult: FAIL - Some annotated images could not be found.")
```

```python
# Annotation–Image Mapping(라벨링–이미지 매핑) 정보 검색

mapping_keywords = [
    "image_01",
    "image_02",
    "image_03",
    "image_04",
    "image_05",
    "image_06",
    "image_07",
    "image_08"
]

search_extensions = {".csv", ".txt", ".md"}

mapping_candidates = []

for path in project_root.rglob("*"):
    if path.is_file() and path.suffix.lower() in search_extensions:
        try:
            text = path.read_text(encoding="utf-8", errors="ignore")

            if any(keyword in text for keyword in mapping_keywords):
                mapping_candidates.append(path)

        except Exception:
            pass

print("=== Annotation–Image Mapping Search ===")
print(f"Candidate Files Found: {len(mapping_candidates)}")

for path in mapping_candidates:
    print("-", path.relative_to(project_root))
```

### QA-04-5. Image Boundary Validation(이미지 경계 검증)

Pilot Image(파일럿 이미지)의 실제 Image Dimension(이미지 크기)을 확인한 결과,
8개 이미지 모두 동일한 **1600 × 900 pixels** 크기로 구성되어 있음을 확인하였다.

따라서 각 Bounding Box(바운딩 박스)의 좌표가 이미지의 유효 범위 내에 존재하는지 검증한다. 

### 검증 기준

- `x_min >= 0`
- `y_min >= 0`
- `x_max <= 1600`
- `y_max <= 900`
- Bounding Box(바운딩 박스)의 모든 좌표가 이미지 경계 내에 존재해야 함

```python
# Image Boundary(이미지 경계) 검증

IMAGE_WIDTH = 1600
IMAGE_HEIGHT = 900

left_boundary_issue = df["x_min"] < 0
top_boundary_issue = df["y_min"] < 0
right_boundary_issue = df["x_max"] > IMAGE_WIDTH
bottom_boundary_issue = df["y_max"] > IMAGE_HEIGHT

boundary_issues = (
    left_boundary_issue |
    top_boundary_issue |
    right_boundary_issue |
    bottom_boundary_issue
)

print("=== Image Boundary Validation ===")
print(f"Image Size: {IMAGE_WIDTH} x {IMAGE_HEIGHT}")
print(f"Total Bounding Boxes: {len(df)}")

print(f"\nLeft Boundary Issues: {left_boundary_issue.sum()}")
print(f"Top Boundary Issues: {top_boundary_issue.sum()}")
print(f"Right Boundary Issues: {right_boundary_issue.sum()}")
print(f"Bottom Boundary Issues: {bottom_boundary_issue.sum()}")

print(f"\nBounding Boxes Outside Image Boundary: {boundary_issues.sum()}")

if boundary_issues.sum() == 0:
    print("Result: PASS - All bounding boxes are within image boundaries.")
else:
    print("Result: REVIEW - Bounding boxes outside image boundaries detected.")
```

### QA-04-6. Bounding Box Size Validation(바운딩 박스 크기 검증)

Bounding Box(바운딩 박스)의 Width(너비), Height(높이), Area(면적)을 계산하여 
박스 크기의 분포를 확인하고 비정상적으로 작거나 큰 객체가 존재하는지 검토한다. 

작은 Bounding Box(바운딩 박스)가 반드시 라벨링 오류를 의미하는 것이 아니며, 
원거리 객체 또는 Small/Distant(소형·원거리) 객체일 가능성이 있으므로 
크기 이상치는 자동 오류로 판정하지 않고 Review Candidate(검토 후보)로 분류한다. 

### 주요 검증 항목

- Bounding Box Width(바운딩 박스 너비) 분포
- Bounding Box Height(바운딩 박스 높이) 분포
- Bounding Boa Area(바운딩 박스 면적) 분포
- 이미지 전체 면적 대비 Bounding Box Area Ratio(바운딩 박스 면적 비율)
- 비정상적인 크기의 Review Candidate(검토 후보) 식별

```python
# Bounding Box Size(바운딩 박스 크기) 계산

df["bbox_area"] = df["bbox_width"] * df["bbox_height"]

image_area = IMAGE_WIDTH * IMAGE_HEIGHT
df["bbox_area_ratio"] = df["bbox_area"] / image_area

print("=== Bounding Box Size Summary ===")

size_summary = df[
    ["bbox_width", "bbox_height", "bbox_area", "bbox_area_ratio"]
].describe()

display(size_summary)

print("\n=== Extreme Values ===")
print(f"Minimum Width: {df['bbox_width'].min():.2f}")
print(f"Minimum Height: {df['bbox_height'].min():.2f}")
print(f"Minimum Area: {df['bbox_area'].min():.2f}")

print(f"\nMaximum Width: {df['bbox_width'].max():.2f}")
print(f"Maximum Height: {df['bbox_height'].max():.2f}")
print(f"Maximum Area: {df['bbox_area'].max():.2f}")

print(
    f"\nMaximum Image Area Ratio: "
    f"{df['bbox_area_ratio'].max() * 100:.2f}%"
)
```

### QA-04-7. Small/Distant–Bounding Box Size Consistency
(소형·원거리 속성–바운딩 박스 크기 일관성 검증)

Bounding Box Size Validation(바운딩 박스 크기 검증) 결과를
`small_distant` 속성과 비교하여 라벨링 기준의 일관성을 확인한다.

Small/Distant(소형·원거리)로 라벨링된 객체는 일반 객체보다
Bounding Box(바운딩 박스)의 크기가 상대적으로 작을 것으로 예상된다.

단, Bounding Box Size(바운딩 박스 크기)만으로 객체의 실제 거리나
라벨링 오류를 확정할 수 없으므로 본 검증에서는 두 그룹의 크기 분포를
비교하고 잠재적인 Review Candidate(검토 후보)를 식별한다.

### 주요 검증 항목

- `small_distant=True`와 `False`의 객체 수
- 두 그룹의 평균 및 중앙 Bounding Box Area(바운딩 박스 면적)
- 두 그룹의 Bounding Box Area Ratio(바운딩 박스 면적 비율)
- 속성과 Bounding Box Size(바운딩 박스 크기)의 전반적인 일관성

```python
# Small/Distant(소형·원거리)와 Bounding Box Size(바운딩 박스 크기) 비교

size_by_distance = (
    df.groupby("small_distant")
    .agg(
        object_count=("object_id", "count"),
        mean_width=("bbox_width", "mean"),
        median_width=("bbox_width", "median"),
        mean_height=("bbox_height", "mean"),
        median_height=("bbox_height", "median"),
        mean_area=("bbox_area", "mean"),
        median_area=("bbox_area", "median"),
        mean_area_ratio=("bbox_area_ratio", "mean")
    )
    .round(2)
)

print("=== Small/Distant vs Bounding Box Size ===")
display(size_by_distance)
```

### QA-04-8. Size Consistency Review Candidate
(크기 일관성 검토 후보)

Small/Distant(소형·원거리) 속성과 Bounding Box Size(바운딩 박스 크기)의
전체적인 관계는 일관된 것으로 확인되었다.

추가적으로 두 그룹의 Median Area(중앙 면적)를 기준으로 반대 그룹의
일반적인 크기 특성을 보이는 객체를 Review Candidate(검토 후보)로 식별한다.

### Review 기준

- `small_distant=True`이면서 Bounding Box Area(바운딩 박스 면적)가
  `False` 그룹의 Median Area(중앙 면적) 이상인 객체
- `small_distant=False`이면서 Bounding Box Area(바운딩 박스 면적)가
  `True` 그룹의 Median Area(중앙 면적) 이하인 객체

본 기준은 Annotation Error(라벨링 오류)를 자동 판정하기 위한 것이 아니라,
속성 라벨과 객체 크기의 관계가 예외적으로 나타나는 사례를 선별하여
Visual QA(시각적 품질검수) 대상으로 활용하기 위한 것이다.

```python
# 그룹별 Median Area(중앙 면적)
median_area_false = df.loc[
    df["small_distant"] == False, "bbox_area"
].median()

median_area_true = df.loc[
    df["small_distant"] == True, "bbox_area"
].median()

# Review Candidate(검토 후보) 추출
large_but_small_distant = df[
    (df["small_distant"] == True) &
    (df["bbox_area"] >= median_area_false)
].copy()

small_but_not_distant = df[
    (df["small_distant"] == False) &
    (df["bbox_area"] <= median_area_true)
].copy()

print("=== Size Consistency Review Candidates ===")
print(f"False Group Median Area: {median_area_false:.2f}")
print(f"True Group Median Area: {median_area_true:.2f}")

print(
    "\nTrue but >= False Median Area: "
    f"{len(large_but_small_distant)}"
)

print(
    "False but <= True Median Area: "
    f"{len(small_but_not_distant)}"
)

print(
    "\nTotal Review Candidates: "
    f"{len(large_but_small_distant) + len(small_but_not_distant)}"
)
```

```python
# Review Candidate(검토 후보) 상세 확인

review_candidates = pd.concat([
    large_but_small_distant,
    small_but_not_distant
]).copy()

review_columns = [
    "image_name",
    "object_id",
    "class_label",
    "bbox_width",
    "bbox_height",
    "bbox_area",
    "small_distant",
    "occlusion",
    "truncation",
    "ambiguity"
]

review_candidates = (
    review_candidates[review_columns]
    .sort_values("bbox_area")
)

print("=== Size Consistency Review Candidate Details ===")
print(f"Total Candidates: {len(review_candidates)}")

display(review_candidates)
```
