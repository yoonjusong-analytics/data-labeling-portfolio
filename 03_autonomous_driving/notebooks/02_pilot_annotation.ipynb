## PA-01. Pilot Annotation Objective (파일럿 라벨링 목적)

본 Pilot Annotation(파일럿 라벨링)은 전체 Dataset(데이터셋)에 
라벨링을 적용하기 전에 실제 이미지에서 Annotation Guideline(라벨링 가이드라인)이 
일관되게 적용될 수 있는지 검증하는 것을 목적으로 한다. 

주요 점증 항목은 다음과 같다. 

- Object Class(객체 클래스) 분류의 일관성 
- Bounding Box(바운딩 박스)의 위치와 범위
- Occlusion(가림) 및 Truncation(잘림) 처리 기준
- Small / Ambiguous Object(소형·모호 객체) 처리 기준
- Missing / Duplicate Annotation(누락·중복 라벨링) 여부
- 전체 Annotation Consistency(라벨링 일관성)

Pilot Annotation에서 발견된 Issue(이슈)와 Edge Case(예외 사례)는 
이후 Annotation Guideline(라벨링 가이드라인)에 반영하여 Full-scale Annotation(본 라벨링)의 품질과 일관성을 높인다.

### Pilot Annotation(파일럿 라벨링) 목적

본 파일럿 라벨링은 전체 Dataset(데이터셋)에 라벨링을 적용하기 전에
실제 이미지에서 라벨링 기주이 일관되게 적용되는지 검증하는 것을 
목적으로 한다. 

주요 검증 항목은 다음과 같다.

- Object Class(객체 클래스) 분류의 일관성 
- Bounding Box(바운딩 박스) 위치와 범위
- Occlusion(가림) 및 Truncation(잘림) 처리
- Small/ Ambiguous Object(소형·모호 객체) 처리 기준
- 객체 누락 및 중복 라벨링 여부
- 전체적인 라벨링 일관성 

파일럿에서 발견된 이슈와 Edge Case(예외 사례)는 이후 
Annotation Guideline(라벨링 가이드라인)에 반영한다. 

## PA-02. Pilot Sample Selection(파일럿 샘플 선정)

### Objective(목적)

다양한 주행 환경에서 Annotation Guideline(라벨링 가이드라인)의 
적용 가능성을 검증하기 위해 소규모이면서 다양한 특성을 포함하는
Pilot Sample(파일럿 샘플)을 선정한다. 

단순한 장면만을 선택하지 않고 Annotation Difficulty(라벨링 난이도)와 
잠재먹인 Edge Case(예외 사례)를 고려하여 샘플을 구성한다. 

### Sample Selection Criteria(샘플 선정 기준)

Pilot Sample은 다음과 같은 조건을 고려하여 선정한다.

| Category(구분) | Description(설명) |
|---|---|
| Simple Scene(단순 장면) | 객체 수가 적고 명확하게 식별 가능한 장면 |
| Multiple Objects(다중 객체) | 하나의 장면에 여러 Target Object(대상 객체)가 존재 |
| Small Objects(소형 객체) | 이미지에서 작은 영역을 차지하는 객체 |
| Occlusion(가림) | 다른 객체에 의해 일부가 가려진 객체 |
| Truncation(잘림) | 이미지 경계를 벗어나 일부만 보이는 객체 |
| Dense Scene(밀집 장면) | 여러 객체가 겹치거나 가까이 위치한 장면 |
| Ambiguous Object(모호 객체) | Class(클래스) 판단이 어려운 객체 |
| Different Distances(다양한 거리) | 근거리·중거리·원거리 객체가 포함된 장면 |

### Selection Principle(선정 원칙)

Pilot Sample은 전체적으로 다음 조건을 포함하도록 구성한다.

- 명확하고 단순한 Annotation Case(라벨링 사례)
- 중간 수준의 Annotation Difficulty(라벨링 난이도)
- 복잡하거나 모호한 Scene(장면)
- Guideline Decision(가이드라인 판단)이 필요한 Edge Case(예외 사례)


최종 Pilot Sample Size(파일럿 샘플 수)는 실제 Dataset Structure
(데이터셋 구조)와 Candidate Review(후보 검토) 결과를 기반으로 확정한다. 

## PA-03. Pilot Candidate Review(파일럿 후보 검토)

### PA-03-1 Review Objective(검토 목적)

최종 Pilot Sample(파일럿 샘플)을 확정하기 전에 Candidate Image
(후보 이미지)를 검토하여 다양한 Annotation Condition
(라벨링 조건)이 포함되어 있는지 확인한다.

Candidate Review(후보 검토)에서는 이미지의 시각적 복잡성과
기존 Annotation Characteristic(라벨 특성)을 함께 고려한다.

주요 검토 기준은 다음과 같다.

- Labeled Object Count(라벨 객체 수)
- Object Class Diversity(객체 클래스 다양성)
- Object Size and Visibility(객체 크기 및 가시성)
- Occlusion / Overlap(가림 및 객체 겹침)
- Image Boundary Object(이미지 경계 객체)
- Scene Density(장면 밀집도)
- Ambiguous Object(모호 객체)

Candidate Image는 검토 결과를 기반으로 다음 세 가지
Preliminary Difficulty(예비 난이도)로 분류한다.

- **Easy(쉬움)**: 객체가 명확하고 겹침이나 모호성이 낮은 장면
- **Medium(보통)**: 다중 객체, 소형 객체 또는 일부 Occlusion이 존재하는 장면
- **Hard(어려움)**: Dense Scene, 심한 Occlusion, Truncation 또는 Class Ambiguity가 존재하는 장면

#### PA-03-9A. Metadata File Inspection(메타데이터 파일 확인)

nuScenes의 Metadata Directory(메타데이터 디렉토리)를 확인하여
Pilot Candidate(파일럿 후보) 분석에 필요한 Metadata File(메타데이터 파일)이
정상적으로 존재하는지 검증한다. 

특히 Candidate Image와 Object Annotation(객체 라벨)을 연결하기 위해 
`sample`, `sample_data`, `sample_annotation`, `category` 정보를 
중점적으로 확인한다. 

```python
#### PA-03-09A. Inspect nuScenes metadata files

metadata_dir = raw_dir / "v1.0-mini"

metadata_files = sorted([
    f for f in metadata_dir.iterdir()
    if f.is_file()
])

print("=== NUSCENES METADATA FILES ===")
print(f"Metadata directory: {metadata_dir}")
print(f"Total files: {len(metadata_files)}\n")

for idx, file in enumerate(metadata_files, start=1):
    print(f"{idx:02d} | {file.name}")
```

#### PA-03-9B. Metadata Loading and Validation(메타데이터 로딩 및 검증)

Pilot Candidate(파일럿 후보)와 Object Annotation(객체 라벨)을
연결하기 위해 주요 Meta File(메타데이터 파일)을 로드하고 
기본 구조를 검증한다. 

본 단계에서는 다음 Metadata를 확인한다. 

- `sample.json`: Sample(샘플) 단위 정보
- `sample_data.json`: 이미지와 Sensor Data(센서 데이터) 정보
- `category.json`: Object Category(객체 카테고리) 정보
- `object_ann.json`: Object Annotation(객체 라벨) 정보

각 Metadata의 Record Count(레코드 수), Column(컬럼) 및 주요 
Identifier(식별자)를 확인하여 이후 Candidate별 Annotation 정보를 
연결할 수 있는지 검증한다. 

```python
#### PA-03-9B. Load and validate metadata

import json
import pandas as pd

metadata_names = [
    "sample.json",
    "sample_data.json",
    "category.json",
    "object_ann.json"
]

metadata_dfs = {}

for filename in metadata_names:
    file_path = metadata_dir / filename

    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    
    df_meta = pd.DataFrame(data)
    metadata_dfs[filename] = df_meta

    print(f"=== {filename}===")
    print(f"Shape: {df_meta.shape}")
    print(f"Columns: {list(df_meta.columns)}")
    print()


```

#### PA-03-9C. Candidate Annotation Linkage(후보 이미지 라벨 연결)

CAM_FRONT Candidate Image(후보 이미지)를 `sample_data` Metadata와
연결한 후, `object_ann` 및 `category` 정보를 결합하여 각 이미지에 
포함된 Object Annotation(객체 라벨)을 확인한다. 

Metadata Join(메타데이터 조인)은 다음 Key(키)를 기준으로 수행한다. 

- `sample_data.token` ↔ `object_ann.sample_data_token`
- `object_ann.category_token` ↔ `category.token`

이를 통해 각 Candidate별 Annotated Object Count(라벨 객체 수),
Object Category(객체 카테고리), Category Diversity(카테고리 다양성)를 
정략적으로 분석하다. 

```python
#### PA-09-9C. Link CAM_FRONT candidates with annotation metadata 

sample_df = metadata_dfs["sample.json"]
sample_data_df = metadata_dfs["sample_data.json"]
category_df = metadata_dfs["category.json"]
object_ann_df = metadata_dfs["object_ann.json"]

# Candidate filenames
candidate_filenames = [
    img_path.name for img_path in valid_front_images
]

# Match CAM_FRONT candidate images in sample_data
candidate_sample_data = sample_data_df[
    sample_data_df["filename"]
    .apply(lambda x: Path(x).name)
    .isin(candidate_filenames)
].copy()

print("Matched candidate images:")
print(candidate_sample_data.shape)

display(
    candidate_sample_data[
        ["token", "sample_token", "filename", "width", "height", "is_key_frame"]
    ]
)
```

## PA_04. Annotation Guideline Setup(라벨링 가이드라인 설정)

Pilot Annotation(파일럿 라벨링)을 수행하기 전에 Object Annotation(객체 라벨링)의 
범위와 판단 기준을 정의한다. 

본 Guideline(가이드라인)은 작업자 간 판단 차이를 최소화하고 Annotation Consistency(라벨링 일관성)를 
확보하기 위한 기준으로 사용한다. 

Pilot Annotation 과정에서 발견되는 Edge Case(예외 사례)는 별도로 기록하고,
QA Review(QA 검토) 결과에 따라 Guideline을 업데이트한다.

### PA-04-1. Annotation Scope(라벨링 범위)

본 Pilot Annotation은 선정된 8개의 CAM_FRONT Key Frame(키 프레임)을 
대상으로 2D Bounding Box(2D 바운딩 박스) 기반 Object Annotation(객체 라벨)을 
수행한다. 

Annotation Scope는 자율주행 환경에서 직접적인 인지 대상이 되는 
Road User(도로 이용 객체)와 주요 Roadside Object(도로 주변 객체)를 
중심으로 정의한다. 

기본 원칙은 다음과 같다. 

- 실제 이미지에서 시각적으로 확인 가능한 객체만 라벨링한다. 
- 각 객체에는 하나의 Class Label(클래스 라벨)을 부여한다. 
- 객체의 Visible Boundary(가시 경계)를 기준으로 가능한 한 밀착된 
  Bounding Box(바운딩 박스)를 생성한다. 
- Occlusion(가림), Truncation(잘림), Small Object(소형 객체)와 같이 
  판단이 필요한 사례는 별도의 세부 규칙을 적용한다. 
- Class를 신뢰성 있게 판단할 수 없는 객체는 임의로 추정하지 않고 
  Ambiguous Case(모호 사례)로 처리한다. 
- 동일한 기준을 모든 Pilot Sample에 일관되게 적용한다. 

본 Pilot의 목적은 Ground Truth(정답 라벨)을 그대로 복제하는 것이 아니라,
정의한 Guideline에 따라 독립적인 Annotation을 수행한 후 기준 Metadata와
비교하여 Annotation Quality(라벨 품질)를 평가하는 것이다. 

### PA-04-2. Target Class Definition(대상 클래스 정의)

Pilot Annotation(파일럿 라벨링)에서는 자율주행 환경에서 중요한 
Road User(도로 이용 객체)와 Roadside Object(도로 주변 객체)를 
중심으로 Target Class(대상 클래스)를 정의한다. 

전체 nuScenes Category(카테고리)를 그대로 사용하는 대신, 
Pilot Sample에서 실젤 관찰 가능하고 시각적으로 일관된 판단이 
가능한 객체를 중심으로 Class Taxonomy(클래스 체계)를 구성한다. 

| Class | Definition(정의) | Annotation Criteria(라벨링 기준) |
|---|---|---|
| Car | 일반 승용 차량 | Sedan, SUV 등 일반 승용 차량 |
| Truck | 화물 운송 차량 | 일반 승용차보다 큰 화물 차량 |
| Construction Vehicle | 건설 작업용 차량 | 공사 현장에서 사용되는 특수 차량 |
| Trailer | 견인되는 화물 구조물 | 차량에 의해 견인되는 Trailer |
| Pedestrian | 보행자 | 도로 및 주변에서 식별 가능한 사람 |
| Bicycle | 자전거 | 식별 가능한 Bicycle 및 탑승 객체 |
| Traffic Cone | 교통 통제용 Cone | 도로 또는 공사 구역의 Traffic Cone |
| Barrier | 교통 통제용 Barrier | 차량 또는 보행자 이동을 제한하는 Barrier |

#### Class Assignment Principles(클래스 부여 원칙)

- 하나의 객체에는 하나의 Target Class만 부여한다. 
- 객체의 시각적 특징이 충분히 확인되는 경우에만 Class를 확정한다. 
- 부분적으로 가려진 객체도 Class를 신뢰성 있게 판단할 수 있으면
  해당 Class로 라벨리한다.
- 서로 다른 객체가 겹쳐 있더라도 개별 객체로 식별 가능하면 
  각각 독립적인 Bounding Box(바운딩 박스)를 생성한다. 
- Class 판단이 어려운 경우 임의로 추정하지 않고 
  Ambiguous Case(모호 사례)로 기록한다. 
- Target Class에 포함되지 않는 객체는 기본적으로 Annotation 
  대상에서 제외한다. 

#### Special Case: Bicycle and Rider(자전거와 탑승자)

- Bicycle(자전거)과 Rider(탑승자)가 모두 명확하게 식별되는 경우 각각 별도의 객체로 라벨링한다. 
- Bicycle은 `Bicycle`, 탑승자는 `Pedestrian`으로 분류한다. 
- 객체가 심하게 가려져 독립적으로 식별할 수 없는 경우에는 Ambiguous Case(모호 사례)로 기록한다. 

### PA-04-3. Bounding Box Rules(바운딩 박스 규칙)

모든 Target Object(대상 객체)는 이미지에서 확인 가능한 
Visible Boundary(가시 경계)를 기준으로 2D Bounding Box(2D 바운딩 박스)를
생성한다. 

Bounding Box는 객체의 위치와 크기를 최대한 정확하게 표현하면서 불필요한 
Background(배경)가 포함되지 않도록 설정한다. 

#### General Bounding Box Principles(기본 바운딩 박스 원칙)

- Bounding Box는 객체의 Visible Boundary에 최대한 밀착하여 생성한다. 
- 객체의 가장 왼쪽, 오른쪽, 위쪽, 아래쪽 Visible Pixel(가시 픽셀)을 
  기준으로 Box Boundary(박스 경계)를 설정한다. 
- 객체 주변의 불필요한 Background(배경)는 가능한 한 포함하지 않는다. 
- Shadow(그림자), Reflection(반사), 빛 번짐은 객체 영역에 포함하지 않는다. 
- 서로 겹쳐 있는 객체도 개별 객체로 식별할 수 있으면 각각 별도의 Bounding Box를
  생성한다. 
- 하나의 객체를 여러 개의 Bounding Box로 분리하지 않는다. 
- 서로 다른 객체를 하나의 Bounding Box로 묶지 않는다. 
- 객체의 보이지 않는 영역을 임의로 추정하여 Bounding Box를 확장하지 않는다. 

#### Bounding Box Edge Cases(바운딩 박스 예외 사례)

**1. Occluded Object(가려진 객체)**
다른 객체나 구조물에 의해 일부가 가려진 경우 실제로 확인 가능한 
Visible Area(가시 영역)를 기준으로 Bounding Box를 생성한다.
가려진 영역의 크기나 형태를 임의로 추정하지 않는다. 

**2. Truncated Object(잘린 객체)**
객체가 Image Boundary(이미지 경계)를 벗어나는 경우 이미지 내부에서 
확인 가능한 영역까지만 Bounding Box를 생성한다. 

**3. Small / Distant Object(소형·원거리 객체)**
객체가 작거나 멀리 있더라도 Target Class를 신뢰성 있게 식별할 수 있고 
객체의 경계를 구분할 수 있다면 Bounding Box를 생성한다. 

**4. Overlapping Object(겹친 객체)**
두 개 이상의 객체가 서로 겹쳐 있더라도 각각 독립적인 객체로 
식별할 수 있으면 개별 Bounding Box를 생성한다. 

**5. Ambiguous Boundary(모호한 경계)**
객체의 일부 경계가 불명확한 경우 확인 가능한 시각적 정보를 기준으로 
가장 보수적인 Bounding Box를 생성하고 해당 사례를 Review 대상에 기록한다. 

#### Bounding Box Coordinate Format(바운딩 박스 좌표 형식)

Bounding Box 좌표는 다음 형식을 사용한다. 

`[x_min, y_min, x_max, y_max]`

- `x_min`: 객체의 왼쪽 경계
- `y_min`: 객체의 위쪽 경계
- `x_max`: 객체의 오른쪽 경계
- `y_max`: 객체의 아래쪽 경계

모든 좌표는 Image Coordinate System(이미지 좌표계)을 기준으로 
Pixel(픽셀) 단위로 기록한다. 

### PA-04-4. Occlusion Rules(가림 처리 규칙)

Occlusion(가림)은 Target Object(대생 객체)의 일부가 다른 객체,
도로 시설물 또는 주변 구조물에 의해 보이지 않는 상태를 의미한다. 

Occluded Object(가려진 객체)는 객체의 Visible Area(가시 영역)를 
기준으로 Class(클래스)와 Boundary(경계)를 신뢰성 잇게 판단할 수 있는 경우 
Annotation(라벨링) 대상에 포함한다. 

보이지 않는 영역의 형태나 크기는 임의로 추정하지 않으며, 
Bounding Box(바운딩 박스)는 실제로 확인 가능한 Visible Boundary(가시 경계)를 
기준으로 생성한다. 

#### Occlusion Level(가림 수준)

| Level | Definition(정의) | Annotation Rule(라벨링 규칙) |
|---|---|---|
| None | 객체의 주요 영역이 가려지지 않음 | 정상적으로 라벨링 |
| Partial | 객체 일부가 가려졌지만 Class와 경계를 식별할 수 있음 | Visible Area를 기준으로 라벨링 |
| Heavy | 객체 대부분이 가려져 Class 또는 경계 판단이 어려움 | 신뢰성 있게 식별 가능한 경우에만 라벨링 |

#### Annotation Principles(라벨링 원칙)

- 다른 차량이나 객체에 일부가 가려져 있어도 독립적인 객체로 식별할 수 
  있으면 별도로 라벨링한다. 
- Partial Occlusion(부분 가림)의 경우 확인 가능한 Visible Boundary를 
  기준으로 Bounding Box를 생성한다. 
- Heavy Occlusion(심한 가림)의 경우 Class와 객체 경계를 신뢰성 있게 
  판단할 수 있을 때만 라벨링한다. 
- 객체 존재는 확인되지만 Class 판단이 어려운 경우 임의로 Class를 
  추정하지 않고 Ambiguous Case(모호 사례)로 기록한다.
- 가려진 부분을 추정하여 Bounding Box를 확장하지 않는다. 
- Occlusion 여부는 Annotation Record(라벨 레코드)에 별도 Attribute(속성)로
  기록하여 이후 QA Review(QA 검토)에 활용한다. 

#### Occlusion Decision Examples(가림 판단 예시)

**Occlusion으로 처리하는 경우**

- 앞 차량이 뒤 차량의 일부를 가리는 경우 
- Pedestrian(보행자)이 차량이나 도로 시설물 뒤에 일부 가려진 경우
- 나무, Barrier(방벽), Traffic Cone(트래픽 콘) 등의 구조물이 Target Object 일부를 가리는 경우 

**Occlusion으로 처리하지 않는 경우**

- 객체가 Image Boundary(이미지 경계)를 벗어난 경우
 → Truncation(잘림)으로 처리 
- 객체가 멀리 있어 작게 보이는 경우
 → Small / Distant Object(소형·원거리 객체)로 처리
- 어둠이나 빛 때문에 객체가 선명하지 않는 경우 
 → Visibility / Ambiguous Case(가시성·모호사례)로 검토

### PA-04-5. Truncation Rules(잘림 처리 규칙)

Truncation(잘림)은 Target Object(대상 객체)의 일부가 Image Boundary(이미지 경계)를 
벗어나 이미지내부에 전체 형태가 기록되지 않은 상태를 의미한다.

Truncated Object(잘린 객체)는 이미지 내부에서 확인 가능한 영역을 
기준으로 Class(클래스)와 객체 경계를 신뢰성 있게 판단할 수 있는 
경우 Annotation(라벨링) 대상에 포함한다. 

Bounding Box(바운딩 박스)는 이미지 내부에서 실제로 확인 가능한 
Visible Boundary(가시 경계)까지만 생성하며, 이미지 밖의 객체 형태나 
크기를 임의로 추정하지 않는다. 

#### Truncation Level(잘림 수준)

| Level | Definition(정의) | Annotation Rule(라벨링 규칙) |
|---|---|---|
| None | 객체 전체가 Image Boundary 내부에 존재 | 정상적으로 라벨링 |
| Partial | 객체 일부가 Image Boundary 밖으로 잘렸지만 Class와 경계를 식별할 수 있음 | 이미지 내부 Visible Area를 기준으로 라벨링 |
| Heavy | 객체 대부분이 Image Boundary 밖에 있어 Class 또는 경계 판단이 어려움 | 신뢰성 있게 식별 가능한 경우에만 라벨링 |

#### Annotation Principles(라벨링 원칙)

- Image Boundary에 걸쳐 있는 객체도 독립적인 Target Object로
  식별할 수 있으면 라벨링한다. 
- Bounding Box는 이미지 내부에서 확인 가능한 Visible Boundary까지만 
  설정한다. 
- 이미지 밖으로 이어지는 객체의 실제 크기나 형태를 추정하여 Bounding Box를 확장하지 않는다. 
- Heavy Truncation(심한 잘림) 상태에서도 Class와 Visible Boundary를 
  신뢰성 있게 판단할 수 있는 경우에는 라벨링한다. 
- 객체 존재는 확인되지만 Class를 신뢰성 있게 판단할 수 없는 경우 
  임의로 추정하지 않고 Ambiguous Case(모호사례)로 기록한다. 
- Truncation 여부는 Annotation Record(라벨 레코드)에 별도 Attribute(속성)로
  기록하여 이후 QA Review(QA 검토)에 활용한다. 


#### Truncation Decision Examples(잘림 판단 예시)

**Truncation으로 처리하는 경우**

- 차량의 앞부분 또는 뒷부분이 이미지 좌우 경계를 벗어난 경우
- Pedestrian(보행자)의 신체 일부가 이미지 경계 밖에 있는 경우 
- Bicycle(자전거) 또는 기타 Target Object의 일부가 이미지 
  상단이나 하단 경계를 벗어난 경우

**Truncation으로 처리하지 않는 경우**

- 다른 차량이나 객체가 Target Object 일부를 가리는 경우
 → Occlusion(가림)으로 처리
- 객체가 멀리 있어 작게 보이는 경우
 → Small / Distant Object(소형·원거리 객체)로 처리 
- 객체의 경계가 어둠, 빛 또는 낮은 해상도로 불명확한 경우
 → Ambiguous Case(모호 사례)로 검토


### PA-04-6. Small / Distant Object Rules(소형·원거리 객체 처리 규칙)

Small / Distant Object(소형·원거리 객체)는 이미지에서 차지하는 영역이 
작거나 Camera(카메라)로부터 멀리 위치하여 세부 특징과 경계가 제한적으로 
보이는 Target Object(대상 객체)를 의미한다.

객체의 크기만을 기분으로 Annotation(라벨링) 여부를 결정하지 않으며, 
Class(클래스)와 Visible Boundary(가시 경계)를 신뢰성 있게 판단할 수
있는지를 우선적인 기준으로 사용한다. 

#### Annotation Decision(라벨링 판단 기준)

| Condition(조건) | Class 식별 | Boundary 식별 | Annotation Rule(라벨링 규칙) |
|---|---|---|---|
| Clear | 가능 | 가능 | 라벨링 |
| Small but Identifiable | 가능 | 대체로 가능 | 라벨링 |
| Boundary Unclear | 가능 | 어려움 | Ambiguous Case로 검토 |
| Class Unclear | 어려움 | 가능 또는 불가능 | 라벨링 제외 또는 Ambiguous Case로 기록 |
| Extremely Small | 어려움 | 어려움 | 라벨링 제외 |

#### Annotation Principles(라벨링 원칙)

- 객체가 작거나 멀리 있더라도 Target Class를 신뢰성 있게 식별할 수 
  있고 Visible Boundary를 구분할 수 있으면 라벨링한다. 
- 객체 크기만을 이유로 자동 제외하지 않는다. 
- Class는 식별 가능하지만 정확한 Boundary 판단이 어려운 경우 
  Ambiguous Case(모호 사례)로 기록하여 QA Review(QA 검토) 대상으로
  분류한다. 
- 몇 개의 Pixel(픽셀)만으로 표현되어 Class와 Boundary를 신뢰성 있게
  판단할 수 없는 객체는 라벨링 하지 않는다. 
- 주변 객체의 형태나 위치를 근거로 보이지 않는 Class를 임의로 
  추정하지 않는다.
- Small / Distant Object 여부는 Annotation Record(라벨 레코드)에 
  Attribute(속성)로 기록하여 이후 QA 및 Error Analysis(오류 분석)에
  활용한다. 

#### Minimum Size Policy(최소 크기 기준)

본 Pilot Annotation에서는 사전에 고정된 Minimum Pixel Threshold(최소 픽셀 임계값)를 
적용하지 않는다. 

이는 동일한 Pixel Size라도 Object Class, Contrast(대비), Occlusion(가림),
Image Quality(이미지 품질)에 따라 식별 가능성이 달라질 수 있기 때문이다. 

Pilot Annotation과 QA Review 결과를 분석한 후 Small Object에서
반복적인 Annotation Inconsistency(라벨링 불일치)가 확인되는 경우,
Minimum Bounding Box Size(최소 바운딩 박스 크기)와 같은 정량적 
Threshold(임계값)의 도임 여부를 검토한다. 

### PA-04-7. Ambiguous Object Rules(모호 객체 처리 규칙)

Ambiguous Object(모호객체)는 이미지에서 객체의 존재는 확인되지만
Class(클래스), Boundary(경계) 또는 Annotation 대상 여부를 신뢰성 있게 
판단하기 어려운 객체를 의미한다. 

모호한 객체는 주변 상황이나 작업자의 추측을 이용하여 임의로 
Class를 결정하지 않는다. 

판단이 불확실한 사례는 Ambiguous Case(모호 사례)로 기록하고 
QA Review(QA 검토)를 통해 최종 판단한다. 

#### Ambiguity Type(모호성 유형)

| Type | Definition(정의) | Example(예시) | Action(처리) |
|---|---|---|---|
| Class Ambiguity | 객체는 확인되지만 Class 구분이 어려움 | 원거리 Car / Truck 구분 불가 | QA Review |
| Boundary Ambiguity | Class는 알 수 있지만 정확한 경계가 불명확 | 어둠 또는 심한 가림 | QA Review |
| Object Ambiguity | 실제 객체인지 판단하기 어려움 | 작은 물체, 반사 또는 그림자 | 라벨링 보류 |
| Target Ambiguity | 객체는 명확하지만 Target Class 대상인지 불분명 | 미정의 Roadside Object | QA Review |

#### Annotation Principles(라벨링 원칙)

- 충분한 시각적 근거 없이 Class를 추정하지 않는다. 
- 주변 객체나 Scene Context(장면 맥락)만을 근거로 Class를 결정하지 않는다. 
- Class를 확정할 수 없으면, 가장 비슷한 Class를 임의로 선택하지 않는다.
- Boundary가 불명확한 경우 객체의 보이지 않는 영역을 추정하여 Bounding Box(바운딩 박스)를
  확장하지 않는다. 
- 객체 존재 자체가 불확실한 경우 Bounding Box를 생성하지 않고 Ambiguous Case로 기록한다. 
- Target Class 여부가 불분명한 객체는 임의로 포함하거나 제외하지 않고 QA Review 대상으로 기록한다. 
- 동일한 유형의 Ambiguous Case가 반복되면 Annotation Guideline(라벨링 가이드라인)의
  추가 또는 수정 필요성을 검토한다. 

#### Ambiguous Case Recording(모호 사례 기록)

Ambiguous Case가 발견되면 다음 정보를 기록한다. 

- `image_name`: 해당 이미지 파일명
- `object_id`: 이미지 내 임시 객체 ID
- `ambiguity_type`: 모호성 유형
- `candidate_class`: 판단 가능한 경우 예상 Class
- `reason`: 판단이 어려운 이유
- `review_status`: QA Review 상태 
- `final_decision`: QA 이후 최종 판단 

예시:

| object_id | ambiguity_type | candidate_class | reason | review_status | final_decision |
|---|---|---|---|---|---|
| OBJ_01 | Class Ambiguity | Car / Truck | Small and distant | Pending | - |
| OBJ_02 | Boundary Ambiguity | Pedestrian | Heavy occlusion | Pending | - |

### PA-04-8. Ignore / Exclusion Rules(제외 객체 처리 규칙)

Ignore / Exclusion Rule(제외 규칙)은 이미지에 존재하는 객체 중 
Pilot Annotation의 Target Class(대상 클래스)에 포함되지 않거나, 
신뢰성 있는 Annotation(라벨링)이 어려운 객체를 의도적으로 라벨링 대상에서 제외하기 위한 기준이다. 

제외된 객체는 Missing Annotation(누락 라벨)과 구분하며, 
필요한 경우 Exclusion Reason(제외 사유)을 기록한ㄷ. 

#### Exclusion Criteria(제외 기준)

| Condition(조건) | Example(예시) | Action(처리) |
|---|---|---|
| Non-target Class | Target Class에 정의되지 않은 객체 | 제외 |
| Extremely Small Object | Class와 Boundary 식별이 불가능한 원거리 객체 | 제외 |
| Unidentifiable Object | 객체 존재 또는 Class를 신뢰성 있게 판단할 수 없음 | 제외 또는 Ambiguous Case |
| Reflection / Shadow | 유리·거울의 반사 또는 객체의 그림자 | 제외 |
| Static Background | 건물, 도로, 나무, 하늘 등의 배경 요소 | 제외 |
| Surface Annotation | 차선, 도로 표면 등 Surface 영역 | 제외 |
| Image Artifact | Blur, Glare 등으로 생성된 비실제 객체 | 제외 |

#### Exclusion Principles(제외 원칙)

- PA-04-2에서 정의한 Target Class에 포함되지 않는 객체는 
  기본적으로 라벨링하지 않는다. 
- Class와 Visible Boundary(가시 경계)를 모두 신뢰성 있게 판단할 수 
  없는 극소형 객체는 라벨링하지 않는다. 
- Reflection(반사), Shadow(그림자), Glare(빛 번짐) 등을 실제 
  객체로 판단하여 Bounding Box(바운딩 박스)를 생성하지 않는다. 
- 건물, 도로, 하늘, 식생 등 Static Background(정적 배경)는 
  Object Annotation(객체 라벨링) 대상에서 제외한다. 
- Lane(차선), Road Surface(도로 표면) 등 Surface Annotation(표면 라벨링) 
  대상은 이번 Pilot Scope에 포함하지 않는다. 
- 단순히 Occlusion(가림) 또는 Truncation(잘림)이 있다는 이유만으로 
  객체를 제외하지 않는다. Class와 Visible Boundary를 신뢰성 있게 
  판단할 수 잇으면 라벨링한다. 
- 제외 여부 자체가 불확실한 경우 즉시 제외하지 않고 
  Ambiguous Case(모호 사례)로 기록하여 QA Review(QA 검토)를 수행한다. 

#### Ignore vs. Ambiguous Decision(제외와 모호 사례 판단)

`Ignore`는 Guideline(가이드라인)에 따라 라벨링 대상이 아니라는 판단이 
명확한 경우에 사용한다. 

`Ambiguous`는 객체가 라벨링 대상일 가능성이 있지만 현재 이미지 정보만으로 
Class, Boundary 또는 Target 여부를 확정하기 어려운 경우에 사용한다. 

따라서 판단 순서는 다음과 같다. 

Target Class 여부 확인 
→ Class / Boundary 식별 가능성 확인 
→ 명확하면 Annotation
→ 불확실하면 Ambiguous
→ 명확한 제외 조건에 해당하면 Ignore

### PA-04-9. Annotation Schema(라벨 스키마)

Pilot Annotation(파일럿 라벨링) 결과는 객체 단위의 Structured Data(구조화 데이터)로 저장한다. 

하나의 Row(행)는 하나의 Annotated Object(라벨링 객체)를 의미하며,
Object Class(객체 클래스), Bounding Box(바운딩 박스), Occlusion(가림),Truncation(잘림) 및 
QA 관련 정보를 함께 기록한다. 

#### Annotation Schema Definition(라벨 스키마 정의)

| Column | Type | Description(설명) |
|---|---|---|
| image_name | string | 원본 이미지 파일명 |
| object_id | string | 이미지 내 객체 고유 ID |
| class_label | string | Target Class(대상 클래스) |
| x_min | int | Bounding Box 왼쪽 좌표 |
| y_min | int | Bounding Box 위쪽 좌표 |
| x_max | int | Bounding Box 오른쪽 좌표 |
| y_max | int | Bounding Box 아래쪽 좌표 |
| occlusion | string | None / Partial / Heavy |
| truncation | string | None / Partial / Heavy |
| small_distant | boolean | 소형·원거리 객체 여부 |
| ambiguity | string | None 또는 Ambiguity Type |
| notes | string | Edge Case 또는 QA 참고사항 |

#### Data Recording Principles(데이터 기록 원칙)

- 하나의 객체는 하나의 Row로 기록한다. 
- `object_id`는 각 이미지 내에서 중복되지 않도록 생성한다. 
- `class_label`은 PA-04-2에서 정의한 Target Class만 사용한다. 
- Bounding Box 좌표는 `[x_min, y_min, x_max, y_max]`기준으로 기록한다. 
- 모든 Bounding Box 좌표는 Pixel(픽셀) 단위의 정수로 저장한다. 
- `occlusion`과 `truncation`은 서로 독립적으로 기록한다. 
- `small_distant`는 `True` 또는 `False`로 기록한다. 
- 모호성이 없는 객체의 `ambiguity`는 `None`으로 기록한다. 
- 추가 검토가 필요한 사례는 `notes`에 판단 근거를 기록한다. 

#### Object ID Convention(객체 ID 규칙)

`object_id`는 다음 형식을 사용한다. 

`C{candidate_number}_OBJ_{object_number}`

예시: 

- `C01_OBJ_001`
- `C01_OBJ_002`
- `C02_OBJ_001`

### PA-04-10. Guideline Summary(가이드라인 요약)

본 Annotation Guideline(라벨링 가이드라인)은 8개의 CAM_FRONT
Pilot Sample을 대상으로 일관된 2D Object Annotation(2D 객체 라벨링)을
수행하기 위한 기준으로 정의하였다.

Guideline의 핵심 목적은 Annotator(작업자)의 주관적 판단을 최소화하고,
동일한 기준을 모든 객체에 적용하여 Annotation Consistency
(라벨링 일관성)를 확보하는 것이다.

#### Core Annotation Rules(핵심 라벨링 규칙)

| Area | Guideline |
|---|---|
| Annotation Scope | 8개의 CAM_FRONT Key Frame을 대상으로 2D Bounding Box 수행 |
| Target Class | Car, Truck, Construction Vehicle, Trailer, Pedestrian, Bicycle, Traffic Cone, Barrier |
| Bounding Box | 실제로 확인 가능한 Visible Boundary에 밀착하여 생성 |
| Occlusion | None / Partial / Heavy로 기록하고 보이지 않는 영역은 추정하지 않음 |
| Truncation | None / Partial / Heavy로 기록하고 이미지 내부 영역까지만 라벨링 |
| Small / Distant Object | Class와 Boundary를 신뢰성 있게 식별할 수 있는 경우 라벨링 |
| Ambiguous Object | 임의로 추정하지 않고 Ambiguous Case로 기록 후 QA Review |
| Exclusion | Non-target, Reflection, Shadow, Static Background, Surface Annotation 등은 제외 |
| Coordinate Format | `[x_min, y_min, x_max, y_max]` Pixel Coordinate 사용 |
| Annotation Unit | 하나의 객체를 하나의 Row로 기록 |

#### Annotation Decision Flow(라벨링 판단 흐름)

각 객체는 다음 순서에 따라 판단한다.

1. Target Class(대상 클래스)에 포함되는 객체인지 확인한다.
2. 객체의 Class와 Visible Boundary(가시 경계)를 식별할 수 있는지 확인한다.
3. 식별 가능하면 Bounding Box(바운딩 박스)를 생성한다.
4. Occlusion(가림), Truncation(잘림), Small / Distant 여부를 기록한다.
5. 판단이 불확실하면 임의로 추정하지 않고 Ambiguous Case로 기록한다.
6. 명확한 Exclusion Rule(제외 규칙)에 해당하면 Annotation 대상에서 제외한다.
7. Edge Case(예외 사례)는 `notes`에 기록하여 QA Review에 활용한다.

#### Quality Control Principles(품질 관리 원칙)

Manual Annotation(수동 라벨링)은 Dataset Ground Truth
(데이터셋 정답 라벨)를 확인하지 않은 상태에서 독립적으로 수행한다.

Pilot Annotation 완료 후 Ground Truth와 비교하여 다음 항목을 평가한다.

- Class Agreement(클래스 일치도)
- Bounding Box IoU(바운딩 박스 IoU)
- Missing Annotation(누락 라벨)
- Extra Annotation(추가 라벨)
- Occlusion / Truncation 관련 판단 차이
- Ambiguous Case 및 반복적인 Edge Case

QA 결과에서 반복적인 오류 또는 판단 불일치가 발견되는 경우
Annotation Guideline을 수정하고 Revision(개정) 내용을 기록한다.

이를 통해 다음과 같은 반복적인 Annotation Quality Improvement
(라벨 품질 개선) Process를 구성한다.

`Guideline → Manual Annotation → QA Review → Error Analysis → Guideline Revision`
