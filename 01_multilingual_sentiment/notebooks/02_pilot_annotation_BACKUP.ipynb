## PA-08. Korean Pilot Finding & Guideline Issues

The Korean pilot annotation identified several ambiguous cases that require 
clarification in the next version of the annotation guideline.

### 1. Neutral vs. Unclear

The distinction between `Neutral` and `Unclear` requires clearer rules.

- `Neutral`: The meaning of the text is understandable, but no clear positive or negative evaluation is expressed.
- `Unclear`: The sentiment cannot reliably determined because of 
insufficient context, incomplete text, or ambiguous meaning.

**Pilot examples:**
-`"다이애나를 평민으로 만든 영화"`→ `Unclear / High`
 - Title only; insufficient information to determine sentiment.

 **Guideline revision candidate:**
 Add an explicit decision rule distinguishing descriptive but interpretable 
 reviews from context-insufficient reviews.

 ---

 ### 2. Confidence for Unclear Labels

 The pilot revealed ambiguity in assigning confidence to `Unclear` labels.

 An annotation may be confidently classified as `Unclear` when the annotator
 is certain that the available text does not provide enough evidence for
 sentiment classification.

 Examples:

 - Title-only review → `Unclear / High`
 - Ambiguous laughter or incomplete expression → `Unclear / Low`

 **Guideline revision candidate:**
 Clarify that confidence represents confidence in the annotation decision, 
 not the strength or clarify of the sentiment itself.

 ### 3. Sentiment Target

 Some reviews express sentiment toward ratings, reviewers, characters,
 or individual scenes rather than toward the movie itself.

 **Pilot examples:**

 - Criticism of rating manipulation
 - Criticism of other reviewers
 - Criticism of a specific movie scene

 **Guideline revision candidate:**
 Define whether sentiment should reflect: 

 1. the overall emotional polarity expressed in the review, or
 2. sentiment specifically directed toward the reviewed product/movie.

 A consistent target rule should be established before main annotation.

 ---

 ### 4. Incomplete and Context-Poor Reviews 

 Short or incomplete reviews frequently require additional interpretation.

 Examples include:

 - incomplete sentences
 - isolated movie titles 
 - quotations without sufficient context
 - short expression followed only by `ㅋㅋ` or `ㅎㅎ`

 **Guideline revision candidate:**
 Add more examples for distinguishing `Neutral` from `Unclear` in
 short and incomplete reviews.

 ### 5. Korean Internet Expressions

 Korean laughter markers such as `ㅋㅋ` and `ㅎㅎ` do not independently indicate positive sentiment.

 Their interpretation depends on surrounding context and may represent: 

 - amusement 
 - sarcasm
 - ridicule
 - embarrassment
 - no clear sentiment

 **Guideline revision candidate:**
 Expand the Korean-specific section with examples of laughter markers 
 used in positive, negative, sarcastic, and unclear contexts.

---

## PA-13 English Pilot Findings & Guideline Issues

The English pilot annotation identified several cases that require
additional clarification in the next version of the annotation guideline.

### 1. Dominant Sentiment in Reviews with Both Positive and Negative Comments

Several reviews contained both positive and negative expressions.
The number of positive or negative phrases alone was not sufficient
to determine the overall sentiment.

The annotation decision should consider:

- the main purpose of the review,
- the importance of the product feature being evaluated,
- the review title and conclusion,
- whether the positive or negative issue dominates the overall experience.

**Example**

> "Ring is beautiful but not sterling silver... Very disappointed."

Although the review contains a positive comment about appearance,
the product-quality problem and explicit disappointment dominate.

**Decision:** `Negative / High`

**Guideline issue:**  
Add a clearer dominance rule explaining that sentiment should be
determined by the overall evaluation rather than by counting
positive and negative expressions.

---

### 2. Core Product Function vs. Minor Positive or Negative Features

Some reviews included minor positive comments while reporting failure
of the product's main function.

For example, a product may have good appearance or secondary benefits
while failing to perform its primary purpose.

**Decision principle:**  
Failure of a core product function may outweigh minor positive
features when determining overall sentiment.

**Guideline issue:**  
Clarify how annotators should weigh core functionality against
secondary product attributes.

---

### 3. Product Sentiment vs. Purchase or Fulfillment Experience

Some reviews evaluated the product positively but described a negative
purchase, packaging, delivery, or order-fulfillment experience.

**Example**

> "The product is good but my order was short of a soap."

The product evaluation is positive, while the fulfillment experience
is negative.

**Decision:** `Mixed / Medium`

**Guideline issue:**  
Define whether sentiment annotation should cover:

- the product only, or
- the complete customer experience including delivery, packaging,
  fulfillment, and customer service.

A consistent sentiment target should be established for future
annotation.

---

### 4. Mixed vs. Dominant Sentiment

The English pilot produced one `Mixed` case.

`Mixed` should be used when meaningful positive and negative
evaluations are both present and neither clearly dominates.

However, the presence of words such as "but", "however", or "although"
does not automatically make a review Mixed.

**Example**

> "Works great... However, the temperature gauge fluctuates."

The minor negative issue does not outweigh the overall positive
evaluation.

**Decision:** `Positive / Medium`

**Guideline issue:**  
Add more contrastive examples showing the difference between:

- `Mixed`
- `Positive with minor negative comments`
- `Negative with minor positive comments`

---

### 5. Confidence Reflects Annotation Certainty

The English pilot contained:

- High confidence: 21 (70.0%)
- Medium confidence: 9 (30.0%)
- Low confidence: 0

The absence of Low-confidence cases does not indicate a QA problem.
Confidence should represent the annotator's certainty about the
assigned label rather than sentiment intensity.

**Guideline issue:**  
Explicitly distinguish:

> Confidence in the annotation decision

from

> Strength or intensity of the sentiment.

---

### 6. English Pilot Summary

The English pilot consisted of 30 reviews.

Final sentiment distribution:

- Negative: 21 (70.0%)
- Positive: 8 (26.7%)
- Mixed: 1 (3.3%)
- Neutral: 0
- Unclear: 0

QA identified one additional review case (`Mixed / Medium`), and the
annotation was retained after review.

No missing sentiment labels or confidence values were found.

### Proposed Guideline v0.2 Updates

Based on the English pilot, the next guideline revision should consider:

1. Expanding the dominant-sentiment decision rule.
2. Clarifying how core product functionality should be weighted.
3. Defining the sentiment target: product vs. complete customer experience.
4. Adding clearer Mixed vs. dominant-sentiment examples.
5. Clarifying that confidence represents annotation certainty rather
   than sentiment intensity.

These findings will be combined with the Korean and Chinese pilot
results before updating the guideline from v0.1 to v0.2.

## PA-19 Chinese Pilot Findings & Guideline Issues

The Chinese pilot annotation identified several language-specific
and cross-language issues that should be clarified in the next
version of the annotation guideline.

### 1. Mild Chinese Sentiment Expressions

Some Chinese reviews expressed sentiment using relatively mild
phrases such as:

- 还不错
- 还好
- 可以
- 一般
- 有点

These expressions may indicate weak sentiment intensity, but they
do not necessarily imply low annotation confidence.

**Example**

> "书还不错 纸张也还好 发货到货都快。"

The positive sentiment is mild, but the overall evaluation is
consistently positive.

**Decision:** `Positive / High`

**Guideline issue:**  
Clarify that mild sentiment intensity should not automatically
reduce confidence.

---

### 2. Implicit Negative Sentiment

Several reviews expressed dissatisfaction without using strong
negative adjectives.

Negative sentiment was sometimes communicated through:

- unmet expectations,
- loss of interest,
- rhetorical questions,
- comparison with better alternatives,
- delayed shipping or service failure.

**Example**

> "以为可以媲美东野圭吾，不过是我的一厢情愿。"

The review does not rely on a strong explicit negative word,
but disappointment and unmet expectations are clear.

**Decision:** `Negative / High`

**Guideline issue:**  
Add Chinese examples of implicit negative sentiment and
context-based interpretation.

---

### 3. Dominant Sentiment in Reviews Containing Both Positive and Negative Comments

Several Chinese reviews contained both positive and negative
evaluations.

The presence of words such as:

- 但是
- 不过
- 就是

should not automatically lead to a `Mixed` label.

The annotator should determine which evaluation dominates the
overall review.

**Example**

> "质量好又便宜... 有些功能没什么用... 用久了觉得好舒服。"

Although several minor concerns are mentioned, the final product
evaluation is clearly positive.

**Decision:** `Positive / High`

**Guideline issue:**  
Add contrastive Chinese examples for:

- Positive with minor negative comments
- Negative with minor positive comments
- Mixed sentiment

---

### 4. Product Sentiment vs. Fulfillment and Service Experience

As in the English pilot, several Chinese reviews evaluated the
product positively while expressing dissatisfaction with:

- shipping,
- packaging,
- stock availability,
- customer service,
- order fulfillment.

**Example**

> "鞋是好鞋... 鞋盒都破了，鞋子也刮花了..."

The product itself is evaluated positively, while the purchasing
and delivery experience is strongly negative.

**Decision:** `Negative / High`

**Guideline issue:**  
Define a consistent sentiment target for reviews containing
multiple evaluation targets.

Possible targets include:

- product quality,
- content quality,
- delivery,
- packaging,
- seller service,
- overall customer experience.

---

### 5. Content Quality vs. Product or Edition Quality

Some book reviews distinguished between the quality of the content
and the quality of the physical or digital edition.

**Example**

> "书是很棒了... 但是中间的低级错误太多了。"

The book content is positively evaluated, while the Kindle edition
contains many formatting and text errors.

**Decision:** `Negative / Medium`

**Guideline issue:**  
Clarify how to annotate reviews where content quality and edition
or production quality have different sentiment directions.

---

### 6. Template-like or Reward-driven Reviews

One review contained positive wording but most of the text described
copying generic review text to earn reward points.

**Example**

> "很好... 这段话复制走了... 既能赚积分，还省事..."

**Decision:** `Positive / Medium`

The review contains positive wording, but actual product-specific
evaluation is limited.

**Guideline issue:**  
Add a rule for template-like, spam-like, or reward-driven reviews.

Future versions should clarify when such reviews should remain
Positive or Negative and when they should be labeled `Unclear`
because the text does not contain reliable product evaluation.

---

# PA-20 Guideline v0.2 Revision Plan

The multilingual pilot annotation was conducted on Korean, English,
and Chinese review samples.

Based on the pilot findings and QA results, the following revisions
will be applied to the annotation guideline before the main
annotation stage.

## 1. Correct Label Set

The official label set is:

`Positive | Negative | Neutral | Mixed | Unclear`

`Negative` was omitted from the label-set summary in v0.1 and will
be corrected in v0.2.

---

## 2. Clarify Neutral vs. Unclear

### Neutral

Use `Neutral` when the meaning of the review is understandable,
but the text contains no meaningful positive or negative evaluation.

### Unclear

Use `Unclear` when the sentiment itself cannot be reliably
determined because the text is incomplete, ambiguous,
context-dependent, meaningless, or insufficient.

### Decision Rule

Ask:

**Can the meaning of the review be understood reliably?**

- Yes, but no positive or negative evaluation exists → `Neutral`
- No, the intended sentiment cannot be determined reliably → `Unclear`

Weak sentiment alone must not be labeled `Neutral`.

---

## 3. Expand the Dominant Sentiment Rule

When positive and negative opinions appear together, do not determine
the label by counting positive and negative expressions.

Consider:

1. Main purpose of the review
2. Importance of the evaluated feature
3. Core product or service functionality
4. Review title
5. Final conclusion or recommendation
6. Severity of the reported problem

If one sentiment clearly dominates the overall evaluation,
assign the dominant sentiment.

Use `Mixed` only when both positive and negative evaluations are
meaningful and neither clearly dominates.

The presence of contrastive expressions such as:

- EN: but / however / although
- KO: 하지만 / 그런데 / 다만
- ZH: 但是 / 不过 / 就是

does not automatically indicate `Mixed`.

---

## 4. Define the Sentiment Target

The annotation target will be the **overall customer experience
expressed in the review**, not only the physical product.

Relevant evaluation targets may include:

- Product quality
- Product functionality
- Content quality
- Digital or physical edition quality
- Delivery
- Packaging
- Order fulfillment
- Seller or customer service

When multiple targets have different sentiment directions,
apply the dominant sentiment rule.

If meaningful positive and negative evaluations remain balanced,
use `Mixed`.

---

## 5. Clarify Core Function vs. Secondary Attributes

Failure of a core product or service function may outweigh
minor positive attributes.

For example:

- Attractive design + major functional failure → usually `Negative`
- Minor packaging issue + excellent product performance → usually `Positive`

Annotators should evaluate the importance of each issue rather
than the number of positive or negative statements.

---

## 6. Clarify Confidence Level

Confidence represents:

**certainty in the annotation decision**

and does NOT represent:

**strength or intensity of sentiment**

### High

The annotator can confidently determine the appropriate label,
even when the sentiment expression itself is mild.

### Medium

Some interpretation or comparison between competing signals
is required.

### Low

The intended sentiment or appropriate label remains substantially
ambiguous or highly context-dependent.

An `Unclear` annotation may still receive `High` confidence when
the annotator is highly certain that the available text is
insufficient for sentiment determination.

---

## 7. Expand Language-Specific Rules

### Korean

Add guidance for:

- ㅋㅋ / ㅎㅎ / ㅠㅠ
- Internet slang and abbreviations
- Double negatives
- Honorific sarcasm
- Incomplete or context-poor expressions

Laughter markers alone must not determine sentiment.

### English

Add guidance for:

- Negation
- Understatement
- Sarcasm
- Intensifiers
- Contrastive structures
- Core-function failure vs. minor positive attributes

### Chinese

Add guidance for mild expressions such as:

- 还不错
- 还好
- 还行
- 一般
- 有点

Weak sentiment intensity does not automatically imply
`Neutral` or lower confidence.

Also consider implicit negative sentiment expressed through:

- Unmet expectations
- Rhetorical questions
- Comparison
- Loss of interest
- Service or delivery failure

---

## 8. Add Guidance for Template-like Reviews

Reviews that contain generic, copied, reward-driven, or
template-like text require additional attention.

If reliable sentiment toward the reviewed item or customer
experience is still present, annotate that sentiment.

If the text does not provide enough reliable evaluative
information, consider `Unclear`.

Document borderline cases in `notes`.

---

## 9. Expand Notes Requirements

`notes` are especially recommended for:

- Mixed sentiment
- Unclear cases
- Low confidence
- Sarcasm
- Language-specific expressions
- Dominant-sentiment decisions involving competing signals
- Product vs. fulfillment conflicts
- Multiple sentiment targets
- Template-like or reward-driven reviews

---

## 10. QA Rule Update

QA should include:

- All `Low` confidence cases
- All `Mixed` cases
- All `Unclear` cases
- Missing annotation fields
- Inconsistent application of dominant sentiment
- Sentiment-target conflicts
- Recurring language-specific ambiguity
- Cross-language consistency checks

Class distributions should be monitored but should not be
artificially balanced during annotation.
