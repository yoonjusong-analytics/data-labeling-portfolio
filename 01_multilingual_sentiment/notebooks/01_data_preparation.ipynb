```python
from datasets import load_dataset

english_dataset = load_dataset(
    "goosmanlei/amazon_reviews_multi",
    "en"
)
```

```python
import pandas as pd

df_ko = pd.read_csv(
    "../data/raw/ratings_train.txt",
    sep="\t"
)

print(df_ko.shape)
print(df_ko.columns.tolist())

df_ko.head()
```

```python
df_ko.isnull().sum()
```

```python
df_ko = df_ko.dropna(subset=["document"]).copy()
```

```python
df_ko["label"].value_counts().sort_index()
```

```python
sample_ko = (
    df_ko
    .groupby("label", group_keys=False)
    .sample(n=150, random_state=42)
    .reset_index(drop=True)
)
```

```python
sample_ko["label"].value_counts().sort_index()
```

```python
sample_ko.shape
```

```python
annotation_ko = sample_ko[
    ["id", "document"]
].copy()

annotation_ko = annotation_ko.rename(
    columns={
        "id": "review_id",
        "document": "review_body"
    }
)

annotation_ko["review_title"] = ""
annotation_ko["language"] = "ko"
annotation_ko["product_category"] = "movie"
annotation_ko["sentiment_label"] = ""
annotation_ko["confidence"] = ""
annotation_ko["notes"] = ""
```

```python
annotation_ko = annotation_ko[
    [
        "review_id",
        "review_title",
        "review_body",
        "language",
        "product_category",
        "sentiment_label",
        "confidence",
        "notes"
    ]
]
```

```python
annotation_ko = (
    annotation_ko
    .sample(frac=1, random_state=42)
    .reset_index(drop=True)
)
```

```python
print(annotation_ko.shape)
print(annotation_ko.columns.tolist())

annotation_ko.head()
```

```python
sample_ko.to_csv(
    "../data/raw/nsmc_ko_ground_truth_300.csv",
    index=False,
    encoding="utf-8-sig"
)
```

```python
annotation_ko.to_csv(
    "../data/annotation/nsmc_ko_annotation_300.csv",
    index=False,
    encoding="utf-8-sig"
)
```
