# Multilingual Sentiment Annotation & QA

## Objective

This notebook performs the main annotation and quality assurance (QA)
process for the multilingual sentiment dataset.

The annotation follows the criteria defined in the annotation guideline
developed during the pilot annotation stage.

### Annotation Labels

- **Polarity:** Positive / Negative
- **Intensity:** Low / Medium / High
- **Languages:** English (EN), Chinese (ZH), Korean (KR)

### Workflow

1. Load the source dataset
2. Select an annotation batch
3. Perform manual annotation
4. Save annotation results
5. Conduct QA review
6. Analyze disagreements and edge cases
7. Update the annotation guideline when necessary

```python
# AQ-02. Import libraries

import pandas as pd
from pathlib import Path
```

```python
# AQ-03. Set project paths

BASE_DIR = Path.cwd().parent

DATA_DIR = BASE_DIR / "data"
RAW_DIR = DATA_DIR / "raw"
PROCESSED_DIR = DATA_DIR / "processed"

print("Base directory:", BASE_DIR)
print("Raw data directory:", RAW_DIR)
print("Processed data directory:", PROCESSED_DIR)
```

```python
# AQ-04. Check available raw data files

for file in RAW_DIR.iterdir():
    print(file.name)
```

```python
# AQ-05. Load English source dataset

EN_FILE = RAW_DIR / "amazon_reviews_en_ground_truth_300.csv"

df_en = pd.read_csv(EN_FILE)

print("Dataset shape:", df_en.shape)
print("\nColumns:")
print(df_en.columns.tolist())

df_en.head()
```

## AQ-06. English Annotation Batch 01

The first main annotation batch contains English reviews #30–49.

To reduce annotation bias, existing star ratings and identifying
metadata are excluded from the annotation view.

The annotator assigns:

- **Polarity:** Positive / Negative
- **Intensity:** Low / Medium / High

```python
# AQ-06. Create English annotation batch #30-49

en_batch_01 = df_en.loc[
    30:49,
    ["review_id", "review_title", "review_body"]
].copy()

en_batch_01["polarity"] = ""
en_batch_01["intensity"] = ""

print("Batch size:", len(en_batch_01))

en_batch_01
```

## AQ-07. Manual Annotation — English Batch 01

Each review is evaluated independently based on its title and full review text.

Existing star ratings are intentionally hidden to minimize annotation bias.

```python
# AQ-07. Display full review text for manual annotation

for idx, row in en_batch_01.iterrows():
    print(f"[{idx}]")
    print(f"Title: {row['review_title']}")
    print(f"Body : {row['review_body']}")
    print("-" * 80)
```

## AQ-08. QA Review — English Batch 01

The manual annotations were independently reviewed using Annotation Guideline v0.2.

The QA process evaluates:

- **Polarity agreement**
- **Intensity agreement**
- **Disagreement reason**
- Potential edge cases for future guideline refinement

> **Key QA principle:**  
> The severity of the reported problem does not necessarily equal the
> intensity of the sentiment expressed in the review.

```python
# AQ-08. Record QA results

qa_results_en_01 = pd.DataFrame({
    "review_id": en_batch_01["review_id"].values,

    "annotator_polarity": ["Negative"] * 20,
    "annotator_intensity": [
        "High", "High", "High", "High", "High",
        "High", "High", "High", "High", "High",
        "High", "High", "High", "High", "High",
        "High", "High", "High", "High", "High"
    ],

    "qa_polarity": ["Negative"] * 20,
    "qa_intensity": [
        "Medium", "High", "Low", "High", "Medium",
        "High", "Medium", "Medium", "High", "High",
        "Medium", "High", "Medium", "Medium", "High",
        "Medium", "Medium", "High", "Medium", "High"
    ]
})

qa_results_en_01["polarity_agreement"] = (
    qa_results_en_01["annotator_polarity"]
    == qa_results_en_01["qa_polarity"]
)

qa_results_en_01["intensity_agreement"] = (
    qa_results_en_01["annotator_intensity"]
    == qa_results_en_01["qa_intensity"]
)

qa_results_en_01
```

## AQ-09. Annotation Agreement Metrics

Agreement rates are calculated separately for polarity and sentiment intensity.

This helps identify whether annotation inconsistencies originate from
sentiment direction or from differences in perceived emotional strength.

```python
# AQ-09. Calculate agreement metrics

polarity_agreement_rate = (
    qa_results_en_01["polarity_agreement"].mean() * 100
)

intensity_agreement_rate = (
    qa_results_en_01["intensity_agreement"].mean() * 100
)

overall_exact_agreement = (
    (
        qa_results_en_01["polarity_agreement"]
        & qa_results_en_01["intensity_agreement"]
    ).mean() * 100
)

print(f"Polarity agreement : {polarity_agreement_rate:.1f}%")
print(f"Intensity agreement: {intensity_agreement_rate:.1f}%")
print(f"Exact label agreement: {overall_exact_agreement:.1f}%")
```

```python
# Extract intensity disagreements

intensity_disagreements = qa_results_en_01[
    ~qa_results_en_01["intensity_agreement"]
].copy()

print("Number of intensity disagreements:", len(intensity_disagreements))

intensity_disagreements
```

## AQ-10. Disagreement Analysis — English Batch 01

### Key Finding

Polarity agreement reached **100%**, indicating that the annotator
consistently identified the direction of sentiment.

However, intensity agreement was **45%**, with 11 disagreements.
All intensity disagreements resulted from the annotator assigning
**High** intensity where the QA review assigned **Medium or Low**.

### Main Disagreement Pattern

The primary source of disagreement was the distinction between:

- **Severity of the reported problem**
- **Intensity of the sentiment expressed in the text**

A serious product or delivery problem does not automatically indicate
high sentiment intensity.

### Examples

- **#32 — Negative / Low**  
  "The seeds were received quickly but were spilled out into shipping envelope."

  The review reports a negative event but contains little explicit
  emotional language.

- **#42 — Negative / Medium**  
  "I did not receive order."

  Non-delivery is a serious issue, but the reviewer communicates it
  briefly without strongly intensified emotional language.

- **#31 — Negative / High**  
  "Will not buy again. Very disappointed."

  Explicit disappointment and rejection of future purchase provide
  clear evidence of high negative intensity.

### QA Insight

Intensity should primarily reflect the **strength of sentiment expressed
through language**, rather than the objective severity of the underlying
product or service problem.

## AQ-11. Final Label Adjudication — English Batch 01

## AQ-25. Polarity Edge Case Analysis — English Batch 03

During Batch 03, several reviews contained both positive and negative
statements or were expressed in an ambiguous or restrained manner.

The annotator initially used **Mixed, Neutral, or Unclear** for six reviews.
However, the current annotation schema defines polarity as a binary label:

- Positive
- Negative

Therefore, these cases require adjudication based on the review's
**dominant evaluative direction**, rather than the presence of both
positive and negative statements.

### Edge Case 1 — Mixed Sentiment

A review may contain both positive and negative statements without
requiring a separate Mixed label.

The final polarity should reflect the **dominant evaluation or overall
conclusion of the review**.

**Example — #70**

> "Ordered the size I always buy and it’s too small, disappointed in the
> sizing ... other than that it a very nice garment."

Positive evidence:
- "very nice garment"

Negative evidence:
- incorrect sizing
- explicit disappointment

**Adjudication: Negative / Medium**

Although positive product qualities are acknowledged, the main purpose
and conclusion of the review concern dissatisfaction with sizing.

---

**Example — #78**

> "Cool item but very poor packaging"

Positive evidence:
- "really fun idea"
- "would have made a great gift"

Negative evidence:
- packaging arrived torn and dirty
- reviewer could no longer use it as the intended gift

**Adjudication: Negative / Medium**

The product concept is evaluated positively, but the actual purchase
experience failed its intended purpose.

### Edge Case 2 — Neutral vs. Negative

A review should not be labeled Neutral simply because it contains
positive observations alongside negative feedback.

When the reviewer reports clear dissatisfaction, product failure,
or recommends caution or an alternative product, the review has
an evaluative direction.

**Example — #71**

The reviewer notes that the shampoo made the hair feel soft, but repeatedly
states that the product failed its primary purpose and made the original
problem worse.

**Adjudication: Negative / High**

---

**Example — #86**

The reviewer acknowledges affordability and possible limited use cases,
but reports repeated disconnections, product failure, and recommends
considering another product.

**Adjudication: Negative / Medium**

### Edge Case 3 — Short Text vs. Unclear Sentiment

A short review is not automatically unclear.

If the available text contains sufficient evaluative information,
polarity should still be assigned.

**Example — #85**

> "See through- the thin material"

Although extremely short, the statement clearly criticizes the material
quality.

**Adjudication: Negative / Low**

Use Unclear only when the text genuinely lacks enough information to
determine sentiment direction.

## AQ-26. Polarity Decision Rule

When a review contains mixed, balanced, or limited information:

1. Identify all positive and negative evidence.
2. Determine the primary product or service issue being evaluated.
3. Examine the reviewer's overall conclusion or behavioral response.
4. Assign the polarity that best represents the **dominant evaluative direction**.

### Important Distinctions

- Mixed content ≠ Mixed label
- Balanced language ≠ Neutral sentiment
- Short text ≠ Unclear sentiment

Under the current binary annotation schema, Positive or Negative should
be assigned whenever the dominant sentiment can reasonably be determined.

## AQ-45. Key QA Findings

### 1. Intensity Calibration

English Batch 01 showed high polarity consistency but low intensity
agreement (**45%**).

QA analysis identified a systematic tendency to equate the severity of
a reported problem with the strength of sentiment expressed in the text.

Guideline v0.3 therefore clarified:

> **Problem severity ≠ Sentiment intensity**

After refinement, intensity agreement increased to **90% in Batch 02**
and remained at **80% in Batches 03 and 04**.

### 2. Polarity Edge Cases

Batch 03 revealed a second annotation issue involving reviews containing
both positive and negative statements.

The annotator introduced Mixed, Neutral, and Unclear labels, resulting
in polarity agreement of **70%**.

Guideline v0.4 therefore introduced the principle:

> **Mixed content ≠ Mixed label**

Under the binary schema, polarity is determined using the review's
dominant evaluative direction.

Batch 04 subsequently achieved **100% polarity agreement**.

### 3. Overall QA

Across the first **80 English production reviews**:

- Polarity agreement: **92.5%**
- Intensity agreement: **73.8%**
- Exact label agreement: **67.5%**

These results demonstrate an iterative QA workflow in which annotation
disagreements are analyzed, converted into guideline improvements, and
evaluated in subsequent annotation batches.

Because each batch contains only 20 reviews, batch-level improvements
are treated as evidence of improved consistency rather than statistically
conclusive performance changes.

#### Key QA Finding

The primary annotation challenge was not identifying whether a review was

The QA results showed that expression strength alone was insufficient for

needed to be considered.

### QA-04. Guideline Refinement

Based on the interim QA findings, the annotation guideline was refined to
**version 0.3**.

The revision focused primarily on improving consistency in intensity
classification and mixed-sentiment decisions.

#### 1. Issue Severity Added to Intensity Decisions

Intensity should be determined using both:

**Expression Strength + Issue Severity**

A review may therefore receive **High intensity even without strong emotional
language** when the reported issue causes severe practical impact.

Examples include:

- Product cannot perform its core function
- Product becomes unusable shortly after purchase
- Severe physical damage or missing components
- Food or hygiene-related safety concerns
- Major service failure with significant customer impact

#### 2. Mixed-Sentiment Decision Rule

When positive and negative signals appear together, the annotator should
identify the **dominant overall evaluation** rather than simply count positive
and negative expressions.

Priority should be given to:

1. Overall evaluation
2. Final conclusion
3. Primary evaluation target
4. Practical impact of the reported issue

#### 3. Sentiment Reversal Rule

When sentiment changes during a review, the **latest explicit overall
evaluation** should generally receive greater weight.

#### 4. Non-Evaluative Reviews

Reviews containing little or no direct product evaluation should receive the
lowest reasonable intensity under the current binary sentiment schema.

Examples include:

- Expectations before actual use
- Plot or content summaries
- General reflections
- Copied or irrelevant review text

These cases were also identified as candidates for a future
**Neutral / Unclear / Non-evaluative** label if the annotation schema is
expanded.

### QA-05. Post-Guideline Quality Evaluation

Following the interim QA review, the annotation guideline was refined to
version 0.3 and applied to the remaining Chinese reviews.

#### Evaluation Approach

The interim QA dataset (reviews 30–109) preserves both the initial and final
labels, allowing reproducible calculation of agreement metrics.

For later annotation batches, QA was performed continuously during annotation,
but the initial labels were not stored as a separate structured dataset.
Therefore, post-v0.3 batch agreement rates are treated as **process evidence**
rather than directly comparable formal metrics.

This distinction prevents unsupported comparisons between the pre- and
post-guideline annotation stages.

#### Observed Improvements

After introducing guideline v0.3, annotation decisions increasingly followed
explicit rules for:

- Separating **expression strength** from **issue severity**
- Assigning High intensity to severe functional failures even when emotional
  language was limited
- Using the **dominant overall evaluation** for mixed reviews
- Prioritizing the **primary evaluation target**
- Using Low intensity for weak, expected, or non-evaluative sentiment

The later QA process also identified cases where intensity could be
overestimated, particularly when any product defect was automatically treated
as High intensity.

This led to a more balanced interpretation:

> High intensity requires either strong sentiment expression or substantial
> practical impact. The presence of a defect alone does not automatically
> justify a High label.

#### Remaining Ambiguities

Several annotation challenges remained even after guideline refinement:

1. **Mixed reviews**
   - Positive product evaluation combined with complaints about price,
     delivery, packaging, or secondary features.

2. **Non-evaluative reviews**
   - Reviews containing expectations, summaries, reflections, or copied text
     rather than direct product evaluation.

3. **Target ambiguity**
   - Sentiment directed toward the product, seller, delivery service,
     publisher, or platform within the same review.

4. **Weak sentiment under a binary schema**
   - Expressions such as "还行", "还可以", or "一般" sometimes fall close to
     the boundary between weak positive and weak negative sentiment.

#### Post-Evaluation Conclusion

Guideline v0.3 improved the decision framework by converting recurring QA
disagreements into explicit annotation rules.

The most important improvement was not simply increasing agreement, but
establishing a more consistent and explainable decision process for ambiguous
reviews.

A future annotation cycle should preserve both initial and final labels for
every review so that pre- and post-guideline agreement can be measured
directly and statistically compared.

### QA-06. Key Findings

#### 1. Sentiment Polarity Was More Consistent Than Intensity

The interim QA sample achieved **95.0% sentiment agreement**, compared with
**76.25% intensity agreement**.

This indicates that identifying whether a review was broadly positive or
negative was relatively stable, while determining the appropriate degree of
sentiment strength was substantially more challenging.

---

#### 2. Initial Annotation Tended to Underestimate Intensity

Among the **19 intensity corrections**, **15 (78.9%)** were upward adjustments:

- Low → Medium: 9
- Medium → High: 6

This revealed a systematic tendency to underestimate intensity when reviews
described serious product or service problems without strongly emotional
language.

The finding led to the introduction of **issue severity** as an explicit
criterion in guideline v0.3.

---

#### 3. Mixed Sentiment Required Target-Aware Decisions

Most polarity disagreements occurred when positive and negative signals
appeared in the same review.

Reliable decisions required considering:

- Overall evaluation
- Final conclusion
- Primary evaluation target
- Practical impact of the reported issue

This was particularly important when reviewers liked the core product but
criticized delivery, packaging, price, or secondary features.

---

#### 4. High Intensity Required More Than the Presence of a Defect

After emphasizing issue severity, later QA revealed a potential risk of
overcorrecting minor defects toward High intensity.

The refined decision principle became:

**High intensity = strong sentiment expression OR substantial practical impact**

Minor cosmetic defects or limited inconveniences should not automatically
receive a High label.

This refinement helped balance emotional expression and real-world issue
severity.

---

#### 5. The Binary Sentiment Schema Has Limitations

Several reviews contained little direct product evaluation, including:

- Expectations before actual use
- General reflections
- Content summaries
- Copied or irrelevant review text
- Ambiguous expressions such as "还行" and "还可以"

These cases had to be mapped to Positive or Negative under the current binary
schema.

A future version of the annotation schema could introduce labels such as
**Neutral, Mixed, Unclear, or Non-evaluative** to represent these cases more
precisely.

### QA-07. Final Conclusion

The Chinese sentiment annotation task was completed for **300 Amazon reviews**
with a final completion rate of **100%**.

The final dataset passed structural and schema validation with:

- **0 missing sentiment labels**
- **0 missing intensity labels**
- **0 duplicate review IDs**
- **0 invalid label values**

A formal interim QA review of 80 annotations achieved:

- **73.75% exact agreement**
- **95.00% sentiment agreement**
- **76.25% intensity agreement**

The QA process identified intensity classification as the primary source of
annotation inconsistency. Among 19 intensity corrections, **78.9% were upward
adjustments**, revealing a tendency to underestimate severe product or service
issues when emotional language was limited.

These findings were incorporated into guideline v0.3, which introduced clearer
decision rules for **issue severity, mixed sentiment, primary evaluation target,
and sentiment reversal**.

Continued QA also revealed the risk of overestimating intensity when minor
defects were present. The final decision principle therefore balanced
**expression strength with practical impact**, rather than relying on either
criterion alone.

The project demonstrates that annotation quality depends not only on assigning
labels, but also on systematically reviewing disagreements, identifying error
patterns, refining annotation guidelines, and validating the final dataset.

Future iterations should preserve both initial and final labels for every
annotation and consider expanding the binary sentiment schema with
**Neutral, Mixed, Unclear, or Non-evaluative** categories for ambiguous cases.

## PE-09. Guideline Revision Recommendations

Based on the post-evaluation analysis of the Korean annotation dataset,
several guideline refinements were identified to improve consistency and
reduce unnecessary `Unclear` assignments.

### 1. Clarify Sentiment Target Identification

Annotators should first determine whether the expressed sentiment is directed
toward the movie or toward another target such as:

- an actor or character,
- another reviewer,
- audience ratings,
- social or political issues,
- unrelated external subjects.

When the sentiment target cannot be reliably connected to the movie,
`Unclear` should be considered rather than inferring polarity from isolated
positive or negative words.

### 2. Distinguish Evaluation from Description

Statements that only describe a movie, scene, title, actor, or factual
information should not automatically be treated as sentiment.

Examples include:

- plot or title information,
- factual statements about actors or source material,
- quotations without evaluative context,
- purely descriptive observations.

A Positive or Negative label should require a reasonable evaluative signal.

### 3. Handle Short and Context-Poor Expressions Conservatively

Very short expressions, numbers, isolated words, emojis, or incomplete
sentences may not provide enough evidence for reliable polarity annotation.

Examples:

- isolated words,
- numeric-only comments,
- unexplained laughter or symbols,
- fragmented expressions.

When sentiment cannot be inferred without external context, assign `Unclear`.

### 4. Separate Implicit Sentiment from Insufficient Context

Implicit sentiment should be assigned Positive or Negative when contextual
cues within the review provide a reasonably clear evaluation.

`Unclear` should be reserved for cases where interpretation depends heavily
on information outside the text.

This distinction can reduce unnecessary uncertainty while avoiding
unsupported inference.

### 5. Strengthen Mixed-Sentiment Rules

When both positive and negative evaluations appear in the same review,
annotators should identify the dominant overall sentiment.

Use `Unclear` only when:

- positive and negative signals are approximately balanced, and
- no dominant final evaluation can reasonably be determined.

### 6. Introduce an Evidence Threshold for Decisive Labels

Before assigning Positive or Negative, annotators should be able to identify
at least one meaningful textual cue supporting the decision.

However, explicit sentiment words are not always required when the overall
meaning clearly communicates evaluation.

This approach balances conservative annotation with contextual
interpretation.

### Expected Impact

These revisions are expected to:

- reduce unnecessary `Unclear` labels,
- improve consistency in ambiguous Korean expressions,
- strengthen sentiment-target identification,
- improve treatment of implicit and mixed sentiment,
- preserve high reliability for decisive Positive/Negative annotations.

The revised rules will be incorporated into the next version of the
multilingual annotation guideline.
