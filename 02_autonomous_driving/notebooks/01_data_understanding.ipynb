## DU-04. Dataset Structure

nuImages Dataset(데이터셋)은 여러 **Metadata Table(메타데이터 테이블)**이 서로 연결된 구조로 구성되어 있다.

주요 Table의 Record(레코드) 수를 확인하여 Dataset의 전체 구조와 **Object Annotation(객체 라벨링)** 정보가 어떻게 구성되어 있는지 파악한다. 

주요 Table은 다음과 같다. 

- **Sample(샘플)**: Annotation의 기준이 되는 Keyframe(키프레임) 정보
- **Sample Data(샘플 데이터)**: Camera Image(카메라 이미지) 파일 및 Sensor(센서) 정보
- **Object Annotation(객체 라벨링)**: 객체별 2D Bounding Box(2D 경계 상자) 정보
- **Surface Annotation(영역 라벨링)**: 도로 등의 Semantic Mask(시맨틱 마스크) 정보
- **Category(카테고리)**: 객체와 영역의 Class(클래스) 정의 
- **Attribute(속성)**: 객체의 상태 또는 행동 정보


```python
# Check the number of records in major metadata tables 
tables = {
    "sample" : nuim.sample,
    "sample_data": nuim.sample_data,
    "object_ann": nuim.object_ann,
    "surface_ann": nuim.surface_ann,
    "category": nuim.category,
    "attribute":nuim.attribute
}

table_summary = pd.DataFrame({
    "table": tables.keys(),
    "records": [len(table) for table in tables.values()]
})

table_summary
```

- Dataset에는 **25개의 Category (카테고리)**와 **12개의 Attribute(속성)**가 
  정의되어 있으므로, Pilot Annotation(파일럿 라벨링)을 시작하기 전에

## DU-06. Keyframe Analysis 

`sample_data`에는 Annotation의 기준이 되는 **Keyframe(키프레임)**과 
시간적으로 인접한 **Sweep(연속 프레임)** Image가 함께 포함되어 있다.

`is_key_frame` Field(필드)를 기준으로 두 Image 유형의 분포를 확인하고,
Pilot Annotation(파일럿 라벨링)에 사용할 실제 Annotation 대상 Image를
구분한다. 

```python
# Check keyframe distribution
keyframe_counts = (
    sample_data_df["is_key_frame"]
    .value_counts()
    .rename_axis("is_key_frame")
    .reset_index(name="count")
)

keyframe_counts
```

```python
# Extract keyframe images
keyframe_df = sample_data_df[
    sample_data_df["is_key_frame"] == True
].copy()

print("Number of keyframes:", len(keyframe_df))

keyframe_df[
    ["token","sample_token","filename", "width", "height", "timestamp"]
].head()
```

### Key Camera Validation

`sample` Table(테이블)의 `key_camera_token`과 `sample_data` Table의 
`token`을 비교하여 각 Sample(샘플)이 하나의 **Key Camera Image(주요 카메라 이미지)**와 정확하게 연결되는지 확인한다. 

이를 통해 `is_key_frame=True`로 확인된 50개의 keyframe(키프레임)이 
실제 Annotation(라벨링)의 기준 Image(이미지)와 일치하는지 검증한다. 

```python
# Extract keyframe records
keyframe_df = sample_data_df[
    sample_data_df["is_key_frame"] == True
].copy()

print("Number of keyframes:", len(keyframe_df))

keyframe_df[
    ["token", "sample_token", "filename", "width", "height"]
].head()
```

```python
# Match sample key_camera_token with sample_data token
key_camera_df = sample_df.merge(
    sample_data_df,
    left_on="key_camera_token",
    right_on="token",
    how="left",
    suffixes=("_sample", "_image")
)

print("Number of samples:", len(sample_df))
print("Number of matched key camera images:", key_camera_df["filename"].notna().sum())
print(
    "All matched images are keyframes:",
    key_camera_df["is_key_frame"].all()
)
```

- `sample.key_camera_token`과 `sample_data.token`을 연결한 결과, 
  50개의 Sample(샘플)이 모두 50개의 **Key Camera Image(주요 카메라 이미지)**와

- 매칭된 50개의 Image는 모두 `is_key_frame=True`로 확인되어

- 따라서 이후 **Pilot Annotation(파일럿 라벨링)**에서는 전체 650개의

## DU-08. Object Annotation Structure 

nuImages v1.0-mini에 포함된 **Object Annotation(객체 라벨링)**의

각 Annotation이 어떤 Image(이미지), Category(카테고리),

```python
# Convert object annotation metadata to DataFrame
object_ann_df = pd.DataFrame(nuim.object_ann)

print("Object Annotation shape:", object_ann_df.shape)

print("\nObject Annotation columns:")
print(object_ann_df.columns.tolist())
```

```python
object_ann_df.head()
```

## DU-09. Category Analysis 

`object_ann` Table(테이블)의 `category_token`을 
**Category Metadata(카테고리 메타데이터)**와 연결하여
각 Object Annotation(객체 라벨링)이 어떤 Object Class(객체 클래스)에 
해당하는지 확인한다. 

또한 Category별 Annotation 수를 분석하여 
nuImages v1.0-mini에서 자주 등장하는 객체와 상대적으로 적게 등장하는 
객체를 파악하고, 향후 **Pilot Annotation(파일럿 라벨링)**에 사용할 
Class(클래스)를 선정하기 위한 기초 자료로 활용한다. 

```python
# Convert category metadata to DataFrame
category_df = pd.DataFrame(nuim.category)

print("Category shape:", category_df.shape)

print("\nCategory columns:")
print(category_df.columns.tolist())

category_df.head()
```

### Category Mapping

`object_ann` Table(테이블)의 `category_token`과 
`category` Table의 `token`을 연결하여 각 
**Object Annotation(객체 라벨링)**에 실제 
**Category Name(카테고리명)**을 부여한다. 

이를 통해 Token(토큰) 형태로 저장된 Category 정보를 사람이 이해할 수 있는 
Object Class(객체 클래스)로 변한하고, 이후 
**Class Distribution (클래스 분포)** 분석에 활용한다.

```python
# Map category tokens to category names
object_category_df = object_ann_df.merge(
    category_df[["token", "name"]],
    left_on="category_token",
    right_on="token",
    how="left",
    suffixes=("_annotation","_category")
)

print("Object annotations:", len(object_category_df))
print(
    "Matched categories:",
    object_category_df["name"].notna().sum()
)

object_category_df[
    ["bbox", "name", "attribute_tokens", "sample_data_token"]
].head()
```

### Class Distribution 

Category Mapping(카테고리 매핑)이 완료된 Object Annotation을 기준으로 
각 **Object Class(객체 클래스)**의 Annotation 수와 전체에서 차지하는 
비율을 계산한다. 

Class별 데이터 분포를 확인하여 자주 등장하는 객체와 
**Rare Class(희소 클래스)**를 파악하고,
Pilot Annotation(파일럿 라벨링)의 Class 선정 기준으로 활용한다.

```python
# Calculate class distribution
class_distribution = (
    object_category_df["name"]
    .value_counts()
    .rename_axis("category")
    .reset_index(name="count")
)

class_distribution["percentage"] = (
    class_distribution["count"]
    / class_distribution["count"].sum()
    * 100
).round(2)

class_distribution
```

### Observation

- 총 **506개의 Object Annotation(객체 라벨링)**에서 
  **15개의 Object Category(객체 카테고리)**가 실제로 사용되고 있다. 
- `human.pedestrian.adult`가 **189개 (37.35%)**로 가장 높은 비중을 차지하며, 
  `vehicle.car`가 **122개(24.11%)**로 두 번째로 높은 비중을 보인다. 
- `movable_object.barrier`는 **70개 (13.83%)**,
  `movable_object.trafficcone`은 **44개(8.70%)**로 나타났다.
- 상위 4개 Category가 전체 Object Annotation의 약 **84.0%**를 차지하여
  뚜렷한 **Class Imbalance(클래스 불균형)**가 존재한다.
- `vehicle.bus.rigid`, `vehicle.construction`, `vehicle.trailer`등의 
  **Rare Class(희소 클래스)**는 Annotation 수가 매우 적으므로,
  단순 Random Sampling(무작위 추출)만으로 Pilot Image(파일럿 이미지)를 
  선정할 경우 일부 Class가 제외될 가능성이 있다.
- 따라서 Pilot Annotation에서는 주요 Class의 대표성과 Rare Class의 
  포함 여부를 함께 고려한 **Stratified Sampling(층화 추출)** 방식의
  Image 선정이 필요하다.

## DU-10. Attribute Analysis 

nuImages의 **Attribute(속성)**는 Object (객체)의 상태 또는 행동 정보를 
표현한다.

예를 들어 동일한 Vehicle(차량)이나 Pedestrian(보행자)이라도
이동 여부, 정차 상태 또는 행동에 따라 서로 다른 Attribute를 가질 수 있다. 

Attribute Metadata(속성 메타데이터)의 구조와 종류를 확인하고,
Object Annotation(객체 라벨링)로 실제로 사용된 Attribute의 분포를 분석한다. 

```python
# Convert attribute metadata to DataFrame
attribute_df = pd.DataFrame(nuim.attribute)

print("Attribute shape:", attribute_df.shape)

print("\nAttribute columns:")
print(attribute_df.columns.tolist())

attribute_df
```

### Attribute Distribution

`object_ann` Table(테이블)의 `attribute_tokens`는 하나 이상의 
**Attribute Token(속성 토큰)**을 List(리스트) 형태로 저장한다. 

`explode()`를 사용하여 Attribute Token을 개별 Record(레코드)로 변환한 뒤, 
`attribute` Table와 연결하여 실제 **Attribute Name(속성명)**을 확인하낟. 

이를 통해 Object (객체)의 상태와 행동이 Dataset(데이터셋)에서 어떻게 분포하는지 분석하고, Pilot Annotation(파일럿 라벨링)에서
Attribute를 포함할지 판단한다. 

```python
# Expand attribute token lists
object_attribute_df = object_ann_df[
    ["token", "category_token", "attribute_tokens"]
].explode("attribute_tokens")

# Map attribute tokens to attribute names
object_attribute_df = object_attribute_df.merge(
    attribute_df[["token", "name"]],
    left_on="attribute_tokens",
    right_on="token",
    how="left",
    suffixes=("_annotation", "_attribute")
)

print("Object annotations:", len(object_ann_df))
print(
    "Annotations with attributes:",
    object_attribute_df["name"].notna().sum()
)

object_attribute_df[
    ["token_annotation", "attribute_tokens", "name"]
].head(10)
```

```python
# Calculate attribute distribution
attribute_distribution = (
    object_attribute_df["name"]
    .dropna()
    .value_counts()
    .rename_axis("attribute")
    .reset_index(name="count")
)

attribute_distribution["percentage"] = (
    attribute_distribution["count"]
    / attribute_distribution["count"].sum()
    * 100
).round(2)

attribute_distribution
```

```python
# Count annotations without attributes
annotations_without_attribute = object_ann_df[
    object_ann_df["attribute_tokens"].apply(len) == 0
]

print(
    "Annotations without attributes:",
    len(annotations_without_attribute)
)

print(
    "Percentage without attributes:",
    round(
        len(annotations_without_attribute)
        / len(object_ann_df) * 100, 
        2
    ),
    "%"
)
```

### Observation

- 총 506개의 **Object Annotation (객체 라벨링)** 중 
  **388개 (76.68%)**에 Attribute(속성)가 부여되지 있으면,
  **188개 (23.32%)**에는 Attribute가 존재하지 않는다.
- Attribute가 부여된 Annotation 중 `pedestrian.moving`이 
  **100개 (25.77%)**로 가장 높은 비중을 차지한다. 
- Vehicle (차량) 관련 Attribute에서는 `vehicle.parked`가

## DU-11. Bounding Box Analysis 

Object Annotation(객체 라벨링)의 **Bounding Box(경계 상자)** 크기를 분석하여
Dataset(데이터셋)에 포함된 객체의 시각적 규모를 파악한다. 

`bbox`의 `[x_min, y_min, x_max, y_max]` 좌표를 이용하여
각 객체의 **Width(너비), Height(높이), Area(면적)**를 계산한다.

특히 Small Object(소형객체)는 Manual Annotation(수동 라벨링) 과정에서
경계 판단이 어렵고 Bounding Box 위치 오차가 상대적으로 크게 발생할 수 있으므로, 
잠재적 Edge Case(경계사례)를 파악한다.

```python
# Extract bounding box coordinates 
bbox_df = object_category_df.copy()

bbox_df[
    ["x_min", "y_min", "x_max", "y_max"]
] = pd.DataFrame(
    bbox_df["bbox"].tolist(),
    index=bbox_df.index
)

# Calculate bounding box dimensions 
bbox_df["bbox_width"] = bbox_df["x_max"] - bbox_df["x_min"]
bbox_df["bbox_height"] = bbox_df["y_max"] - bbox_df["y_min"]
bbox_df["bbox_area"] = bbox_df["bbox_width"] * bbox_df["bbox_height"]

bbox_df[
    ["name", "bbox", "bbox_width", "bbox_height", "bbox_area"]
].head()
```

```python
# Summary statistics for bounding box size 
bbox_summary = bbox_df[
    ["bbox_width", "bbox_height", "bbox_area"]
].describe().round(2)

bbox_summary
```

### Observation

- 506개의 **Object Annotation(객체 라벨링)**에서 Bounding Box(경계 상자)의
  크기가 매우 다양하게 나타났다.
- Bounding Box Width(경계 상자 너비)의 Median(중앙값)은 **43 px**,
  Height(높이)의 Median은 **61 px**이며, 
  Area(면적)의 Median은 **2,753 px²**이다.
- 반면 평균 Area는 **13,549.81 px²**, Maximum Area(최대면적)는
  **925,200 px²**로 나타나 일부 매우 큰 객체가 포함되어 있다. 
- 최초 Bounding Box는 **7 × 17 px**, Area는 **119 px²**로,
  육안으로 식별하기 어려운 Small Object(소형 객체)도 존재한다.
- 이러한 Bounding Box 크기 차이는 Manual Annotation(수동 라벨링)의 
  난이도와 위치 정확도에 영향을 줄 수 잇으므로,
  Small Object를 Quality Assurance(품질 검증)의 주요
  Edge Case(경계사례)로 고려할 필요가 있다. 
  

## DU-12. Relative Bounding Box Size Analysis 

Bounding Box(경계 상자)의 절대 pixel Area(픽셀 면적)뿐 아니라 
전체 Image Area(이미지 면적)에서 객체가 차지하는 비율을 계산한다.

이를 위해 Object Annotation(객체 라벨링)의 `sample_data_token`과 
Sample Data(샘플 데이터)의 Image Resolution(이미지 해상도)을 연결하고,
각 Bounding Box의 **Relative Area(상대 면적)**를 산출한다.

Relative Area를 활용하면 Image Resolution의 영향을 줄이고
Small Object(소형 객체)를 보다 일관된 기준으로 비교할 수 있다. 

```python
# Add image resolution to bounding box data
bbox_relative_df = bbox_df.merge(
    sample_data_df[["token", "width", "height"]],
    left_on="sample_data_token",
    right_on="token",
    how="left",
    suffixes=("_annotation", "_image")
)

print(
    "Matched image resolutions:",
    bbox_relative_df["width"].notna().sum()
)

print(
    "Total object annotations:",
    len(bbox_relative_df)
)
```

```python
# Calculate relative bounding box area
bbox_relative_df["image_area"] = (
    bbox_relative_df["width"] * bbox_relative_df["height"]
)

bbox_relative_df["relative_area"] = (
    bbox_relative_df["bbox_area"]
    / bbox_relative_df["image_area"]
)

bbox_relative_df["relative_area_pct"] = (
    bbox_relative_df["relative_area"] * 100
)

bbox_relative_df[
    [
        "name",
        "bbox_area",
        "image_area",
        "relative_area_pct"
    ]
].head()
```

```python
# Summary statistics for relative bounding box area
relative_area_summary = (
    bbox_relative_df["relative_area_pct"]
    .describe()
    .round(4)
)

relative_area_summary
```

## DU-13. Category × Bounding Box Size Analysis 

각 **Object Category(객체 카테고리)**별 Bounding Box(경계상자)의 
Relative Area(상대 면적)를 비교하여 Object Class(객체 클래스)에 따른 크기 차이를 분석한다. 

Category별 **Annotation Count(라벨 수)**, 
**Median Relative Area(상대 면적 중앙값)**,
**Mean Relative Area(상대 면적 평균)**를 계산하여
Small Object(소형객체)가 주로 어떤 Category에서 발생하는지 확인한다. 

분석 결과는 Pilot Annotation(파일럿 라벨링)에서
Annotation Difficulty(라벨링 난이도)가 높은 Class를 포함하기 위한 
Sampling Criteria(샘플링 기준)를 설정하는데 활용한다. 


```python
# Calculate bounding box size statistics by category
category_size_summary = (
    bbox_relative_df
    .groupby("name")
    .agg(
        annotation_count=("relative_area_pct", "count"),
        median_area_pct=("relative_area_pct", "median"),
        mean_area_pct=("relative_area_pct", "mean")
    )
    .reset_index()
    .sort_values("median_area_pct")
)

# Round values for readability 
category_size_summary[
    ["median_area_pct", "mean_area_pct"]
] = category_size_summary[
    ["median_area_pct", "mean_area_pct"]
].round(4)

category_size_summary
```

  Median Relative Area(상대 면적 중앙값)가 **0.0456%**로 가장 작게 나타났다. 
  따라서 Traffic Cone(교통 콘)은 Small Object(소형 객체) Annotation의

  높은 출현 빈도와 작은 Object Size(객체 크기)를 동시에 보인다. 
- `vehicle.car`는 **122개의 Annotation**과 **0.2999%**의 Median Relative Area를 
  보여 주요 Vehicle Class(차량 클래스) 중 비교적 안정적인 표본을 제공한다.
- `vehicle.motorcycle`, `vehicle.bicycle`, `vehicle.truck`은

  Pedestrian(보행자)과 Traffic Cone보다 상대적으로 크게 나타난다.
- 일부 Rare Class(희소 클래스)는 Median Relative Area가 크게 나타났지만,
  `vehicle.trailer` 2개, `movable_object.debris` 1개 등 
  Sample Size(표본 수)가 매우 작으므로 일반적인 Object Size 특성으로 
  해석하기에는 한계가 있다. 
- 따라서 Pilot Sampling(파일럿 샘플링)에서는 단순 출현 빈도 뿐 아니라 
  **Class Frequency(클래스 빈도), Object Size(객체 크기), Rare Class Coverage(희소 클래스 포함 여부)**를 
  함게 고려할 필요가 있다.

## DU-15. Class Diversity Analysis 

Key Camera Image(주요 카메라 이미지)별 **Class Diversity(클래스 다양성)**를 
분석하여 여러 종류의 Object Category(객체 카테고리)가 동시에 등장하는

## DU-16. Small Object Annotation Analysis (소형 객체 라벨링 분석)

Object Detection(객체 탐지) 데이터에서 이미지 내 차지하는 면적이 작은 객체는 
Annotation Difficulty(어노테이션 난이도)와 Quality Risk(품질 위험)를 높일 수 있다.

본 단계에서는 Bounding Box Area(경계 상자 면적)를 전체 Image Area(이미지 면적)과
비교하여 Relative Bounding Box Area(상대 경계 상자 면적)를 기준으로 객체 크기를 
분류하고, Small Object Annotation(소형 객체 어노테이션)의 분포를 분석한다. 

이를 통해 다음을 확인한다. 

- Small Object(소형 객체)의 비율
- Object Size Group(객체 크기 그룹)의 분포
- Small Object가 많이 나타나는 Category(범주)
- Annotation QA(어노테이션 품질 검수) 시 우선적으로 확인할 수 있는 잠재적 Risk(위험)

```python
# Create object size groups based on relative bounding box area
def classify_object_size(relative_area_pct):
    if relative_area_pct < 0.1:
        return "Very Small"
    elif relative_area_pct < 0.5:
        return "Small"
    elif relative_area_pct < 2.0:
        return "Medium"
    else:
        return "Large"

bbox_relative_df["size_group"]= (
    bbox_relative_df["relative_area_pct"]
    .apply(classify_object_size)
)

# Calculate object size distribution
size_distribution = (
    bbox_relative_df["size_group"]
    .value_counts()
    .reindex(
        ["Very Small", "Small", "Medium", "Large"],
        fill_value=0
    )
    .rename_axis("size_group")
    .reset_index(name="count")
)

size_distribution["percentage"] = (
    size_distribution["count"]
    / len(bbox_relative_df)
    * 100
).round(2)

size_distribution
```

### Observation

- 전체 **506개의 Object Annotation(객체 라벨링)** 중 **169개(33.40%)**가 Very Small Object(초소형 객체), **196개(38.74%)**가 Small Object(소형 객체)로 분류되었다. 
- 두 그릅을 합하면 **365개(72.14%)**로, 전체 객체의 약 72%가 Image Area(이미지 면적)의 **0.5% 미만**을 차지한다. 
- Medium Object(중형 객체)는 **104개(20.55%)**, Large Object(대형 객체)는 **37개(7.31%)**로 나타난다.
- 이는 본 데이터에서 작은 Bounding Box(경계 상자)가 상당한 비중을 차지하고 있음을 보여주며, Small Object Annotation(소형 객체 라벨링)을 잠재적인 Annotation QA Risk (라벨링 품질)을 판단하지
  않는다.
- 단, 본 Size Group(크기 그룹)은 nuImages의 공식 분류 기준이 아니라 본 프로젝트의 QA 분석을 위해 설정한 Operational Threshold(운영 기준값)이므로, 객체 크기만으로 Annotation Quality(라벨링
  품질)를 판단하지 않는다. 

### DU-16.2 Small Object Distribution by Category (범주별 소형 객체 분포)

Very Small Object(초소형 객체)와 Small Object(소형 객체)를 합쳐
Small-object Risk Group(소형 객체 위험 그룹)으로 정의하고 Category(범주)별
분포를 분석한다.

단순한 Small Object Count(소형 객체 수)뿐만 아니라 각 Category 내에서 
소형 객체가 차지하는 비율을 함께 계산하여, Annotation QA (라벨링 품질 검수)시
우선적으로 검토할 Category를 탐색한다. 

```python
# Define small-object risk group
bbox_relative_df["is_small_object"] = (
    bbox_relative_df["relative_area_pct"] < 0.5
)

# Calculate category-Level small-object statistics 
small_object_by_category = (
    bbox_relative_df
    .groupby("name")
    .agg(
        total_annotations=("name", "size"),
        small_object_count=("is_small_object", "sum")
    )
    .reset_index()
)

small_object_by_category["small_object_pct"] = (
    small_object_by_category["small_object_count"]
    / small_object_by_category["total_annotations"]
    * 100
).round(2)

# Sort by number of small objects
small_object_by_category = (
    small_object_by_category
    .sort_values(
        ["small_object_count", "small_object_pct"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

small_object_by_category
```

### Observation

- `human.pedestrian.adult`는 전체 **819개** 중 **162개(85.71%)**가 Small Object(소형 객체)로 분류되었으며, Small Object Count(소형 객체 수)가 가장 많았다.
- `movable_object.trafficcone`은 **44개 중 42개(95.45%)**가 Small Object로, 충분한 Annotation Count(라벨링 수)를 가진 주요 Category 중 가장 높은 
   Small Object Ratio(소형 객체 비율)를 보였다. 
- `movable_object.barrier` 역시 **70개 중 54개(77.14%)**가 Small Object로 나타나 높은 빈도와 높은 소형 객체 비율을 동시에 보였다. 
- `vehicle.car`는 **122개 중 72개(59.02%)**가 Small Object로 분류되어, Vehicle Category(차량 범주)에서도 원거리 또는 작은 객체가 상당한 비중을 차지하는 것을 확인할 수 있다. 
-  반면 일부 Category는 Small Object Ratio가 높거나 낮더라도 전체 Annotation Count 매우 적기 때문에, 비율만으로 Annotation QA Priority(라벨링 품질 우선순위)를 결정할 때는 Sample
   Size(표본크기)를 함께 고려해야 한다.

### DU-16.4 QA Priority Scoring (품질 검수 우선순위 점수화)

Small Object Ratio(소형 객체 비율)가 높더라도 Sample Size(표본 크기)가 매우 작으면 
전체 Annotation QA(라벨링 품질 검수)에 미치는 영향은 제한적일 수 있다. 

따라서 본 단계에서는 각 Category(범주)의 Small Object Count(소형 객체 수)를 
QA Priority Score(품질 검수 우선순위 점수)로 활용하여, 실제 검수 Workload(작업량)가 
큰 Category를 식별한다. 

또한 Small Object Ratio를 함께 제시하여 다음 두 요소를 동시에 평가한다. 

- **QA Workload(검수 작업량):** 실제 Small Object Annotation의 수 
- **QA Exposure(품질 위험 노출도):** 해당 Category에서 Small Object가 차지하는 비율 

본 점수는 Annotation Error Rate(어노테이션 오류율)를 의미하지 않으며, 
잠재적인 QA Review Priority(품질 검수 우선순위)를 설정하기 위한 분석 지표이다.

```python
# Create QA priority table
qa_priority = small_object_by_category.copy()

# Use small-object count as QA workload-based priority score
qa_priority["qa_priority_score"] = qa_priority["small_object_count"]

# Rank categories by QA priority
qa_priority["qa_priority_rank"] = (
    qa_priority["qa_priority_score"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

# Reorder and sort columns
qa_priority = (
    qa_priority[
        [
            "name",
            "total_annotations",
            "small_object_count",
            "small_object_pct",
            "qa_priority_score",
            "qa_priority_rank"
        ]
    ]
    .sort_values(
        ["qa_priority_rank", "small_object_pct"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

qa_priority
```

### Observation

- QA Priority Score(품질 점수 우선순위 점수)는 각 Category(범주)의 Small Object Count(소형 객체 수)를 기준으로 정의하였다. 
- `human.pedestrian.adult`는 **162개의 Small Object Annotation(소형 객체 라벨링)**을 포함하여 QA Priority Rank(품질 검수 우선순위)
  **1위**로 나타났다.

## DU-17. Small Object QA Candidate Inspection(소형 객체 품질 검수 후보 점검)

DU-16에서는 Relative Bounding Box Area(상대 경계 상자 면적)를 기반으로 
Small Object Annotation(소형 객체 라벨링)을 잠재적인 Annotation QA Risk(라벨링 품질 검수 위험) 
대상으로 식별하였다. 

본 단계에서는 데이터 수준에서 확인한 Small Object Risk(소형 객체 위험)를 
실제 Camera Image(카메라 이미지)와 연결하여 시각적으로 검토한다.

특히 QA Priority(품질 검수 우선순위)가 높은 Category를 중심으로 
Bounding Box(경계 상자)가 매우 작은 Annotation을 추출하여 다음을 확인한다. 

- Small Object Annotation(소형 객체 라벨링)의 실제 이미지 내 크기
- Category(범주)별 잠재적 Annotation Difficulty(라벨링 난이도)
- Bounding Box Placement(경계 상자 위치)의 시각적 검토 가능성
- 후속 Manual QA(수동 품질 검수)에 사용할 Candidate(후보) 선정 

본 단계의 목적은 Ground Truth Error(정답 라벨링 오류)를 자동으로 판정하는 것이 아니라,
데이터 기반으로 Manual Review Priority(수동 검수 우선순위)를 선정하는 것이다. 

```python
priority_categories = [
    "human.pedestrian.adult",
    "vehicle.car",
    "movable_object.barrier",
    "movable_object.trafficcone"
]

# Select small-object annotations from high-priority categories
qa_candidates = (
    bbox_relative_df[
        (bbox_relative_df["name"].isin(priority_categories))
        & (bbox_relative_df["relative_area_pct"] < 0.5)
    ]
    .copy()
)

print("Total QA candidates:", len(qa_candidates))

qa_candidates[
    [
        "name",
        "bbox",
        "bbox_width",
        "bbox_height",
        "relative_area_pct",
        "sample_data_token"
    ]
].head(10)
```

### DU-17.2 Category-based QA Sampling (범주별 품질 검수 표본 추출)

전체 Small Object QA Candidate(소형 객체 품질 검수 후보)를 모두 수동으로 
검토하는 대신, 각 Priority Category(우선 검수 범주)에서 
Relative Bounding Box Area(상대 경계 상자 면적)가 가장 작은 Annotation(라벨링)을 우선적으로 추출한다. 

각 Category에서 가장 작은 객체 **5개씩**, 총 **20개**를 선정하여
Manual QA Sample(수동 품질 검수 표본)을 구성한다.

이러한 Risk-based Sampling(위험 기반 표본 추출)을 무작위 표본 추출보다
잠재적으로 검수가 어려운 객체를 우선적으로 확인하기 위한 방법이다. 

```python
# Select 5 smallest annotations from each priority category
qa_sample = (
    qa_candidates
    .sort_values(["name", "relative_area_pct"])
    .groupby("name", group_keys=False)
    .head(5)
    .reset_index(drop=True)
)

print("Total QA sample:", len(qa_sample))

# Check sample count by category 
qa_sample_summary = (
    qa_sample
    .groupby("name")
    .agg(
        sample_count=("name", "size"),
        min_area_pct=("relative_area_pct", "min"),
        max_area_pct=("relative_area_pct", "max")
    )
    .reset_index()
)

qa_sample_summary[
    ["name", "sample_count","min_area_pct", "max_area_pct"]
]
```

### Observation

- 4개의 Priority Category(우선 검수 범주)에서 각각 5개씩,
  통 **20개의 Manual QA Sample(수동 품질 검수 표본)**을 추출하였다.
- 모든 Category(범주)에서 선정된 객체의 Relative Bounding Box Area(상대 경계 상자 면적)는 
  약 **0.008%~0.031% 범위로, 이미지 전체 면적에 비해 매우 작은 객체들이다. 
- `human.pedestrian.adult`의 최소 Relative Area(상대 면적)는 **0.0083%**로 네 Category 가운데 가장 작은 값을 보였다. 
- `movable_object.barrier`와 `movable_object.trafficcone`에서도 각각 약 **0.0099%**, **0.0104%** 수준의 매우 작은 객체가 확인되었다. 
- 이러한 Extreme Small Object(극소형 객체)는 객체의 경계가 불명확하거나 시각적 정보가 제한적일 수 있으므로 Manual QA(수동 품질 검수)의 
  우선 검토 대상으로 활용할 수 있다. 

### DU-17.3 Visual QA Inspection (시각적 품질 검수)

Risk-based Sampling(위험 기반 표본 추출)으로 선정된 QA Sample 가운데 
각 Priority Category(우선 검수 범주)에서 Relative Bounding Box Area(상대 경계 상자 면적)가 
가장 작은 객체를 하나씩 선택한다. 

선정된 4개의 Representative QA Candidate(대표 품질 검수 후보)를 
원본 Camera Image(카메라 이미지)와 Bounding Box(경계 상자)를 함께 
시각화하여 실제 Annotation Difficulty(라벨링 난이도)를 검토한다.

주요 검토 항목을 다음과 같다. 

- Object Visibility(객체 가시성)
- Bounding Box Placement(경계 상자 배치)
- Object Size(객체 크기)
- Occlusion(가림)
- Category Identifiability(범주 식별 가능성)

본 검토는 Ground Truth Error(정답 오류)를 자동 판정하기 위한 것이 아니라,
실제 Manual QA(수동 품질검수)가 필요한 사례를 탐색하기 위한 과정이다. 

```python
# Select the smallest object from each priority category
representative_qa = (
    qa_sample
    .sort_values("relative_area_pct")
    .groupby("name", group_keys=False)
    .head(1)
    .sort_values("relative_area_pct")
    .reset_index(drop=True)
)

print("Representative QA candidates:", len(representative_qa))

representative_qa[
    [
        "name",
        "bbox",
        "bbox_width",
        "bbox_height",
        "relative_area_pct",
        "sample_data_token"
    ]
]
```

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Create one visualization for each representative QA candidate
for _, row in representative_qa.iterrows():

    # Find image metadata using sample_data_token
    image_info = sample_data_df[
        sample_data_df["token"] == row["sample_data_token"]
    ].iloc[0]

    image_path = dataroot / image_info["filename"]

    # Load image
    image = Image.open(image_path)

    # Bounding box coordinates 
    x_min, y_min, x_max, y_max = row["bbox"]

    bbox_width = x_max - x_min
    bbox_height = y_max - y_min

    # Plot image
    fig, ax = plt.subplots(figsize=(12, 7))
    ax.imshow(image)

    # Draw bounding box
    rectangle = patches.Rectangle(
        (x_min, y_min),
        bbox_width,
        bbox_height,
        fill=False,
        linewidth=3
    )

    ax.add_patch(rectangle)

    # Add annotation Label
    ax.text(
        x_min,
        max(y_min - 10, 10),
        f'{row["name"]} | {row["relative_area_pct"]:.4f}%',
        fontsize=10,
        bbox=dict(facecolor="white", alpha=0.8)
    )

    ax.set_title(
        f'Representative Small Object QA Candidate\n'
        f'{row["name"]} | '
        f'{bbox_width}x{bbox_height} px | '
        f'{row["relative_area_pct"]: .4f}%'
    )

    ax.axis("off")
    plt.tight_layout()
    plt.show()
```

### Observation
- 각 Priority Category(우선 검수 범위)에서 가장 작은 Annotation(라벨링)을 선택하여 
  실제 Camera Image(카메라 이미지)에서 Visual QA Inspection(시각적 품질 검수)을 수행하였다. 
- 대표 객체의 Bounding Box Size(경계 상자 크기)는 **7×17 px에서 19×15 px** 수준으로 매우 작았으며, 
  Relative Bounding Box Area (상대 경계 상자 면적)는 약 **0.0083%~0.0198%**에 불과하였다.
- `human.pedestrian.adult`는 **7×17 px**로가장 작은 사례였으며, 원본 이미지에서 객체의 세부 형태와
  경계를 명확하게 식별하기 어려웠다.
- `movable_object.barrier`와 `movable_object.trafficcone` 역시 주변 Scene(장면)에 비해 매우 작아
  Category Identifiability(범주 식별 가능성)와 Bounding Box Placement(경계 상자 배치)에 대한 
  추가 검토가 필요하다. 
- `vehicle.car`는 다른 사례보다 시각적 식별이 상대적으로 용이하지만, **19×15 px**의 작은 Bounding Box이므로 
  몇 Pixel(픽셀)의 경계 차이도 상대적으로 큰 Annotation Variation(어노테이션 변동)을 만들 수 있다.
- 본 Visual Inspection(시각적 검토)만으로 Ground Truth Error(정답 오류)를 확정할 충분한 근거는 발견되지 않았으며, 
  4개 사례 모두 **Review Candidate(검수후보)**로 분류하였다.
- 결과적으로 Small Object Size(소형 객체 크기)는 자동적인 오류 판정 기준이라기보다

### DU-17.4 QA Review Framework (품질 검수 프레임워크)

Visual QA Inspection(시각적 품질 검수)의 일관성을 높이기 위해 
대표 Small Object QA Candidate(소형 객체 품질 검수 후보)에 적용할 
Review Criteria(검수 기준)를 정의한다. 

본 프로젝트에서 다음 세 가지 QA Status(품질 검수 상태)를 사용한다.

- **Pass(통과):** Object Category(객체 범주)와 Bounding Box Placement
  (경계 상자 배치)가 시각적으로 합리적이며 추가 검토가 필요하지 않는 경우 
- **Review(검토 필요):** Object Size(객체 크기), Occlusion(가림),
  Low Visibility(낮은 가시성) 등으로 인해 추가적인 Human Review(사람의 재검토)가 필요한 경우
- **Potential Issue(잠재적 문제):** Category Mismatch(범주 불일치), Missing Object(객체 누락),
  명확한 Bounding Box Misalignment(경계 상자 위치 오류) 등 Annotation Error(어노테이션 오류)가 의심되는 경우

QA Status는 Ground Truth Error(정답 오류)를 자동으로 확정하기 위한 것이 아니라,
Manual QA(수동 품질 검수) 과정에서 Review Priority(검수 우선순위)를 일관되게 관리하기 위한 
Operational Label(운영 라벨)로 사용한다. 

  

| QA Status(품질 상태) | 주요 판단 기준 | 후속 조치 |
|---|---|---|
| **Pass(통과)** | Category와 Bounding Box가 시각적으로 명확하고 합리적임 | 추가 검토 없음 |
| **Review(검토 필요)** | Small Object, Occlusion, Low Visibility 등으로 판단이 어려움 | 확대 확인 또는 재검토 |
| **Potential Issue(잠재적 문제)** | Category 불일치, 객체 누락, Bounding Box 위치 오류가 의심됨 | 상세 검토 및 Error Log 기록 |

Category-level Risk(범주 수준 위험)를 활용하여, Manual Review(수동 검수)가 필요한 대상을 단계적으로 좁힐 수 있었다.

따라서 실제 Annotation QA(어노테이션 품질 검수)에서는 단순한 Random Sampling(무작위 표본 추출)뿐 아니라 데이터 특성을 
기반으로 한 Risk-based Sampling(위험 기반 표본 추출)을 함께 적용하면, 제한된 Review Resource(검수 지원)를 보다 효율적으로 
배분할 수 있다. 

또한 본 프로젝트에서 정의한 **Pass(통과) / Review(검토 필요) / Potential Issue(잠재적 문제)** Framework(프레임워크)는 
후속 Manual QA에서 (운영 가이드라인)으로 활용할 수 있다.
